```python
import torch

from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_ID
)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.eval()
model.to(device)

print("Model device:", device)
print(sum([p.numel() for p in model.parameters() if p.requires_grad]))

```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">این تمرین برای اجرا روی پردازنده مرکزی طراحی شده است.</p>


<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">مرحله ۴.۲ — پیش‌پردازش</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">کارت مدل برای نام کاربری و آدرس اینترنتی یک روش پیش‌پردازش مشخص پیشنهاد می‌کند.</p>


```python
def preprocess(text):
    new_text = []

    for token in text.split(" "):

        if token.startswith("@") and len(token) > 1:
            token = "@user"

        if token.startswith("http"):
            token = "http"

        new_text.append(token)

    return " ".join(new_text)
```

```python
raw_text = (
    "@sara I really like the new update! "
    "https://example.com"
)

processed_text = preprocess(raw_text)

print("Original:")
print(raw_text)

print("\nProcessed:")
print(processed_text)
```

```python
num_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print("Number of parameters:")
print(num_parameters)
```

```python
def parameter_storage_gib(
    parameters,
    bits
):
    return (
        parameters
        * bits
        / 8
        / 1024**3
    )

for name, bits in [
    ("FP32", 32),
    ("FP16", 16),
    ("INT8", 8),
    ("INT4", 4),
]:
    value = parameter_storage_gib(
        num_parameters,
        bits
    )

    print(name, value)
```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">چرا عدد محاسبه‌شده در جدول را نمی‌توان مستقیماً برابر با کل حافظه مصرف‌شده هنگام استنتاج دانست؟</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">تفاوت بین «فضای نظری پارامترها» و «حافظه کل زمان اجرا» را توضیح دهید.</p>

<h1 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">بخش ۹ — مفهوم کوانتیزه‌سازی</h1>
<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">هدف</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">قرار نیست مدل اصلی را کوانتیزه کنیم.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">می‌خواهیم با یک مثال کوچک ببینیم چگونه تبدیل وزن‌های دقیق‌تر به تعداد محدودی کد باعث تقریب می‌شود.</p>


```python
weights = torch.tensor([
    -0.80,
    -0.60,
     0.10,
     0.15,
     0.20,
     0.70
])

scale = weights.abs().max() / 7

int4_values = (
    torch.round(
        weights / scale
    )
    .clamp(-8, 7)
    .to(torch.int8)
)

reconstructed = (
    int4_values.float()
    * scale
)

absolute_error = torch.abs(
    weights - reconstructed
)

print("Original weights:")
print(weights)

print("\nScale:")
print(scale)

print("\nQuantized codes:")
print(int4_values)

print("\nReconstructed weights:")
print(reconstructed)

print("\nAbsolute error:")
print(absolute_error)
```
