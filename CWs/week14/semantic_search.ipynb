<h2 dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">مرحله ۵ — رابطه <bdi dir="ltr">Dot Product</bdi> و <bdi dir="ltr">Cosine Similarity</bdi></h2>
<p dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">در این پروژه هم <bdi dir="ltr">Document Embedding</bdi> و هم <bdi dir="ltr">Query Embedding</bdi> با <bdi dir="ltr">normalize_embeddings=True</bdi> ساخته می‌شوند.</p>
<p dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">توضیح دهید چرا در این حالت <bdi dir="ltr">L2 norm</bdi> هر دو بردار برابر ۱ است و بنابراین <bdi dir="ltr">Dot Product</bdi> با <bdi dir="ltr">Cosine Similarity</bdi> برابر می‌شود.</p>


وقتی شباهت بین دو بردار رو حساب میکنیم با ضرب نقطه ای ، اگر طول دو بردار متفاوت باشد و اندازه آنها این باعث میشود روی شباهت اثر گذار باشد
برای از بین بردن این تاثیر ضرب دوبردار را تقسیم بر ضرب نورم اقلیدسی آنها میکنیم تا طول آنها بی اثر شودو تابع encoder این نرمال سازی را از قبل انجام داده لذا باعث میشود بخش مخرج حذف شود و نیازی به آن نباشد 

<div dir="ltr" style="text-align:left; direction:ltr; unicode-bidi:isolate;">
\[
\cos(q,d)=\frac{q^T d}{\|q\|_2\|d\|_2}
\]
</div>
<div dir="ltr" style="text-align:left; direction:ltr; unicode-bidi:isolate;">If \(\|q\|_2=1\) and \(\|d\|_2=1\), then:</div>
<div dir="ltr" style="text-align:left; direction:ltr; unicode-bidi:isolate;">\[
\cos(q,d)=q^T d
\]</div>

<h2 dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">مرحله ۶ — پیاده‌سازی <bdi dir="ltr">Semantic Search Engine</bdi></h2>
<h3 dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">سناریو</h3>
<p dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;"><bdi dir="ltr">Document Embedding</bdi>ها قبلاً ساخته شده‌اند و نباید برای هر <bdi dir="ltr">Query</bdi> دوباره ساخته شوند.</p>
<h3 dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">کاری که باید انجام دهید</h3>
<p dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">تابع زیر را کامل کنید.</p>

```python
def semantic_search(
    query,
    documents,
    document_embeddings,
    model,
    top_k=3
):
    query_embedding = model.encode(
        query,
        convert_to_tensor=True,
        normalize_embeddings=True
    )

    scores = document_embeddings @ query_embedding

    # TODO:

    # 1) Select top-k with torch.topk(...)
    top_scores,top_idx = torch.topk(scores,k=min(top_k,len(documents)))
    # 2) Map indices back to documents
    # 3) Return id, title, category, score
    results = []
    for score,idx in zip(top_scores,top_idx):
        s_idx = int(idx.item())

        result = documents[s_idx].copy()
        result["score"] = float(score)

        results.append(result)

    return results

```

<h2 dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">مرحله ۷ — اجرای <bdi dir="ltr">Query</bdi>های ارزیابی</h2>
<p dir="rtl" style="text-align:right; direction:rtl; unicode-bidi:plaintext;">پنج <bdi dir="ltr">Query</bdi> و <bdi dir="ltr">Ground Truth</bdi> از قبل مشخص شده‌اند و نباید براساس خروجی سیستم تغییر کنند.</p>

```python
evaluation_queries = [
    {
        "query": "How do I make a separate Python environment?",
        "relevant_ids": {"python-environment"}
    },
    {
        "query": "Where do I submit the final project?",
        "relevant_ids": {"git-submission"}
    },
    {
        "query": "How can I reduce the memory needed by a local language model?",
        "relevant_ids": {"local-llm"}
    },
    {
        "query": "How can I test a model on unseen examples?",
        "relevant_ids": {"model-evaluation"}
    },
    {
        "query": "What is used to classify images?",
        "relevant_ids": {"cnn"}
    }
]
```

```python
# TODO:
# Run semantic_search(..., top_k=3) for all evaluation queries.
result_semantic = []
for query in evaluation_queries:
    result_semantic.append(semantic_search(
        query = query["query"],
        documents=documents,
        document_embeddings=document_embeddings,
        model=model,
    ))



# For each query print:
# QUERY
# GROUND TRUTH
# TOP-3 RETRIEVED
# Found in Top-3? Yes/No
# Ground-truth rank: 1/2/3/Not found
for i,query in enumerate(evaluation_queries):
    print("==="*10)
    print(f"Query: {query['query']}")
    print(f"Ground truth: {query['relevant_ids']}")

    for result in result_semantic[i]:
        print("-"*10)

        print(f"relevant score: {result['score']}")
        print(f"text: {result['text']}")
        print(f"doc id: {result['id']}")
```
