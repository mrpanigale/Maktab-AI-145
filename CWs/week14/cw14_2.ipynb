<div dir="rtl" style="direction:rtl; text-align:right; line-height:2; font-family:Tahoma, Arial, sans-serif; font-size:16px;"><h2>مأموریت ۲ — اجرای اولین پیش‌بینی</h2><p>حالا که <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Model</span>‏ را بررسی کردیم، ساده‌ترین روش اجرا را با <code dir="ltr" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">pipeline()</code>‏ امتحان کنید. برای هر <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Text</span>‏ مقدار <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Label</span>‏ و <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Score</span>‏ را نمایش دهید.</p></div><div dir="ltr" style="direction:ltr; text-align:left; line-height:1.5; font-family:Arial, sans-serif; margin-top:4px; margin-bottom:12px;"><h3 style="margin-top:0;">Prediction with pipeline()</h3></div>

```python
sentiment = pipeline(
    "sentiment-analysis",
    model=SENTIMENT_MODEL_ID,
)

texts = [
    "The explanation was clear and useful.",
    "The application crashes every time I open it.",
    "The course was interesting but difficult.",
]

results = sentiment(texts)

for text, result in zip(texts, results):
    print("Text:", text)
    print("Label:", result["label"])
    print("Score:", result["score"])
    print("-" * 60)
```

<p><b>پاسخ کوتاه:</b> در یک جمله توضیح دهید <code dir="ltr" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">pipeline()</code>‏ چه کاری را برای ما ساده کرده است.</p>

<div dir="rtl" style="direction:rtl; text-align:right; unicode-bidi:isolate; line-height:2; font-family:Tahoma, Arial, sans-serif; font-size:16px;"><div style="border:1px dashed #999; border-radius:8px; padding:12px; margin:8px 0; min-height:60px;"><b>پاسخ شما:</b><br/><br/></div></div>

<div dir="rtl" style="direction:rtl; text-align:right; line-height:2; font-family:Tahoma, Arial, sans-serif; font-size:16px;"><h2>مأموریت ۳ — بررسی مراحل داخلی اجرا</h2><p>یکی از دانشجوهای <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Bootcamp</span>‏ می‌پرسد: «<span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Model</span>‏ که متن خام را مستقیماً پردازش نمی‌کند؛ پس بین <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Text</span>‏ و <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Label</span>‏ چه اتفاقی می‌افتد؟» این بار همان مسیر را مرحله‌به‌مرحله اجرا کنید.</p></div><div dir="ltr" style="direction:ltr; text-align:left; line-height:1.5; font-family:Arial, sans-serif; margin-top:4px; margin-bottom:12px;"><h3 style="margin-top:0;">Tokenizer → Logits → Probabilities → Label</h3></div>

<div dir="ltr" style="direction:ltr; text-align:left; unicode-bidi:isolate; line-height:1.8; font-family:Arial, sans-serif; font-size:15px;">
<pre>
Raw Text
   ↓
Tokenizer
   ↓
input_ids + attention_mask
   ↓
Model
   ↓
Logits
   ↓
Softmax
   ↓
Probabilities
   ↓
Predicted Class
   ↓
Label
</pre>
</div>

```python
tokenizer = AutoTokenizer.from_pretrained(
    SENTIMENT_MODEL_ID
)

model = AutoModelForSequenceClassification.from_pretrained(
    SENTIMENT_MODEL_ID
)

model.eval()

manual_texts = [
    "The lesson was useful.",
    "The result was disappointing.",
]

batch = tokenizer(
    manual_texts,
    padding=True,
    truncation=True,
    return_tensors="pt",
)

print("input_ids:")
print(batch["input_ids"])
print()

print("attention_mask:")
print(batch["attention_mask"])
print()

print("input_ids shape:", batch["input_ids"].shape)
print("attention_mask shape:", batch["attention_mask"].shape)
```

```python
with torch.no_grad():
    outputs = model(**batch)

logits = outputs.logits
probabilities = torch.softmax(logits, dim=-1)
predicted_ids = probabilities.argmax(dim=-1)

print("Logits:")
print(logits)
print()

print("Probabilities:")
print(probabilities)
print()

for i, predicted_id in enumerate(predicted_ids):
    class_id = int(predicted_id)
    label = model.config.id2label[class_id]

    print("Text:", manual_texts[i])
    print("Predicted Class ID:", class_id)
    print("Final Label:", label)
    print("-" * 60)
```

<div dir="rtl" style="direction:rtl; text-align:right; line-height:2; font-family:Tahoma, Arial, sans-serif; font-size:16px;">
<h2>مأموریت ۵ — تخمین <span dir="ltr" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap;">Weight Memory</span></h2>
<p>
مدل فرضی ما دقیقاً <span dir="ltr" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap;">7,000,000,000 Parameters</span> دارد.
در این سؤال فقط یک تخمین ساده و ایده‌آل از Memory مربوط به Weightها محاسبه می‌کنیم.
</p>
</div>
<div dir="ltr" style="direction:ltr; text-align:left; unicode-bidi:isolate; line-height:1.8; font-family:Arial, sans-serif; font-size:15px;">
<h3>Idealized Raw Weight Memory</h3>
</div>


<div dir="ltr" style="direction:ltr; text-align:left; unicode-bidi:isolate; line-height:1.8; font-family:Arial, sans-serif; font-size:15px;">
<p><b>Formula</b></p>
<p>
\[
\text{Idealized Raw Weight Memory}
\approx
\text{Parameters}
\times
\frac{\text{Bits per Parameter}}{8}
\]
</p>
</div>

```python
def weight_memory_gib(parameters, bits):
    return parameters * bits / 8 / 1024**3

parameters = 7_000_000_000

representations = {
    "FP32": 32,
    "FP16": 16,
    "INT8": 8,
    "INT4": 4,
}

rows = []

for name, bits in representations.items():
    memory_gib = weight_memory_gib(parameters, bits)

    rows.append({
        "representation": name,
        "bits_per_parameter": bits,
        "idealized_raw_weight_memory_gib": round(memory_gib, 2),
        "below_8_gib": memory_gib < GPU_VRAM_GIB,
    })

memory_table = pd.DataFrame(rows)
memory_table
```

<div dir="rtl" style="direction:rtl; text-align:right; line-height:2; font-family:Tahoma, Arial, sans-serif; font-size:16px;"><h2>مأموریت ۶ — بررسی اثر Quantization بر Weightها</h2><p>یکی از دانشجوها می‌پرسد: «آیا <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Quantization</span>‏ فقط همان مقدار <span dir="ltr" lang="en" style="direction:ltr; unicode-bidi:isolate; display:inline-block; white-space:nowrap; text-align:left;">Weight</span>‏ قبلی را بدون تغییر در فضای کمتری ذخیره می‌کند؟» برای بررسی، مثال آموزشی زیر را اجرا کنید.</p></div><div dir="ltr" style="direction:ltr; text-align:left; line-height:1.5; font-family:Arial, sans-serif; margin-top:4px; margin-bottom:12px;"><h3 style="margin-top:0;">Quantization</h3></div>

```python
weights = torch.tensor([
    -0.80,
    -0.60,
     0.10,
     0.15,
     0.20,
     0.70,
])

scale = weights.abs().max() / 7

quantized = (
    torch.round(weights / scale)
    .clamp(-8, 7)
    .to(torch.int8)
)

reconstructed = quantized.float() * scale

absolute_error = (
    weights - reconstructed
).abs()

quantization_table = pd.DataFrame({
    "original": weights.tolist(),
    "quantized_code": quantized.tolist(),
    "reconstructed": reconstructed.tolist(),
    "absolute_error": absolute_error.tolist(),
})

quantization_table
```
