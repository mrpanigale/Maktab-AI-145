```python
import torch

from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_ID
)

model.eval()

print("Model device:", next(model.parameters()).device)
```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">این تمرین برای اجرا روی پردازنده مرکزی طراحی شده است.</p>


<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">مرحله ۴.۲ — پیش‌پردازش</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">کارت مدل برای نام کاربری و آدرس اینترنتی یک روش پیش‌پردازش مشخص پیشنهاد می‌کند.</p>


```python
def preprocess(text):
    new_text = []

    for token in text.split(" "):

        if token.startswith("@") and len(token) > 1:
            token = "@user"

        if token.startswith("http"):
            token = "http"

        new_text.append(token)

    return " ".join(new_text)
```

```python
raw_text = (
    "@sara I really like the new update! "
    "https://example.com"
)

processed_text = preprocess(raw_text)

print("Original:")
print(raw_text)

print("\nProcessed:")
print(processed_text)
```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;"><strong>ب)</strong> آدرس اینترنتی چه تغییری کرده است؟</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;"><strong>ج)</strong> چرا هنگام استفاده از یک مدل ازپیش‌آموزش‌دیده بهتر است پیش‌پردازش توصیه‌شده برای همان مدل را بررسی و رعایت کنیم؟</p>

```python
with torch.no_grad():
    outputs = model(**inputs)

logits = outputs.logits

print("Logits:")
print(logits)

print("\nLogits shape:")
print(logits.shape)
```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;"><strong>الف)</strong> بُعد اول به چه چیزی مربوط است؟</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;"><strong>ب)</strong> بُعد دوم چه ارتباطی با تعداد برچسب‌های مدل دارد؟</p>

<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">هدف</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">می‌خواهیم ارتباط بین تعداد پارامترها، تعداد بیت برای نمایش هر پارامتر و فضای نظری موردنیاز برای ذخیره آن‌ها را بررسی کنیم.</p>


```python
num_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print("Number of parameters:")
print(num_parameters)
```

```python
def parameter_storage_gib(
    parameters,
    bits
):
    return (
        parameters
        * bits
        / 8
        / 1024**3
    )

for name, bits in [
    ("FP32", 32),
    ("FP16", 16),
    ("INT8", 8),
    ("INT4", 4),
]:
    value = parameter_storage_gib(
        num_parameters,
        bits
    )

    print(name, value)
```

<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">چرا عدد محاسبه‌شده در جدول را نمی‌توان مستقیماً برابر با کل حافظه مصرف‌شده هنگام استنتاج دانست؟</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">تفاوت بین «فضای نظری پارامترها» و «حافظه کل زمان اجرا» را توضیح دهید.</p>

<h1 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">بخش ۹ — مفهوم کوانتیزه‌سازی</h1>
<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">هدف</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">قرار نیست مدل اصلی را کوانتیزه کنیم.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">می‌خواهیم با یک مثال کوچک ببینیم چگونه تبدیل وزن‌های دقیق‌تر به تعداد محدودی کد باعث تقریب می‌شود.</p>


```python
weights = torch.tensor([
    -0.80,
    -0.60,
     0.10,
     0.15,
     0.20,
     0.70
])

scale = weights.abs().max() / 7

int4_values = (
    torch.round(
        weights / scale
    )
    .clamp(-8, 7)
    .to(torch.int8)
)

reconstructed = (
    int4_values.float()
    * scale
)

absolute_error = torch.abs(
    weights - reconstructed
)

print("Original weights:")
print(weights)

print("\nScale:")
print(scale)

print("\nQuantized codes:")
print(int4_values)

print("\nReconstructed weights:")
print(reconstructed)

print("\nAbsolute error:")
print(absolute_error)
```

<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">سؤال ۱۵</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">فرض کنید فقط برچسب نهایی «مثبت» را ببینید.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">در حداکثر ۸ مرحله توضیح دهید از لحظه ورود متن خام تا تولید این برچسب چه اتفاق‌هایی افتاده است.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">در پاسخ باید جایگاه این مفاهیم مشخص باشد:</p>
<ul dir="rtl" style="direction:rtl;text-align:right;padding-right:24px;padding-left:0;margin-right:0;">
<li dir="rtl" style="direction:rtl;text-align:right;">پیش‌پردازش</li>
<li dir="rtl" style="direction:rtl;text-align:right;">توکن‌سازی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">شناسه‌های ورودی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">ماسک توجه</li>
<li dir="rtl" style="direction:rtl;text-align:right;">مدل</li>
<li dir="rtl" style="direction:rtl;text-align:right;">لاجیت‌ها</li>
<li dir="rtl" style="direction:rtl;text-align:right;">سافت‌مکس</li>
<li dir="rtl" style="direction:rtl;text-align:right;">شناسه کلاس و برچسب</li>
</ul>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">سپس در یک جمله توضیح دهید کوانتیزه‌سازی وزن‌ها در ساده‌ترین شکل چه تغییری در نحوه نمایش وزن‌های مدل ایجاد می‌کند.</p>
<h3 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">پاسخ شما</h3>


<h1 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">موارد تحویلی</h1>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">فایل نهایی باید شامل موارد زیر باشد:</p>
<ol dir="rtl" style="direction:rtl;text-align:right;padding-right:24px;padding-left:0;margin-right:0;">
<li dir="rtl" style="direction:rtl;text-align:right;">اطلاعات کارت مدل</li>
<li dir="rtl" style="direction:rtl;text-align:right;">فراداده</li>
<li dir="rtl" style="direction:rtl;text-align:right;">پیکربندی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">نگاشت شناسه کلاس به برچسب</li>
<li dir="rtl" style="direction:rtl;text-align:right;">پیش‌پردازش</li>
<li dir="rtl" style="direction:rtl;text-align:right;">توکن‌سازی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">شناسه‌های ورودی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">ماسک توجه</li>
<li dir="rtl" style="direction:rtl;text-align:right;">لاجیت‌ها</li>
<li dir="rtl" style="direction:rtl;text-align:right;">خروجی سافت‌مکس</li>
<li dir="rtl" style="direction:rtl;text-align:right;">کلاس و برچسب پیش‌بینی‌شده</li>
<li dir="rtl" style="direction:rtl;text-align:right;">آزمایش پَدینگ</li>
<li dir="rtl" style="direction:rtl;text-align:right;">آزمایش کوتاه‌سازی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">ذخیره و بارگذاری محلی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">تعداد پارامترهای مدل</li>
<li dir="rtl" style="direction:rtl;text-align:right;">جدول فضای ذخیره‌سازی نظری</li>
<li dir="rtl" style="direction:rtl;text-align:right;">مثال کوانتیزه‌سازی</li>
<li dir="rtl" style="direction:rtl;text-align:right;">پاسخ سؤال پایانی</li>
</ol>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">نیازی به تصویر از صفحه نیست.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">نیازی به آموزش یا تنظیم دقیق مدل نیست.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">نیازی به محاسبه دقت کلی مدل نیست.</p>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">نیازی به اجرای کوانتیزه‌سازی واقعی روی مدل اصلی نیست.</p>
<h2 dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;">هدف نهایی</h2>
<p dir="rtl" style="direction:rtl;text-align:right;width:100%;margin-right:0;line-height:1.9;">در پایان باید بتوانید توضیح دهید یک مدل آماده از مخزن هاگینگ‌فیس تا تولید پیش‌بینی چه مسیر محاسباتی‌ای را طی می‌کند و تعداد بیت‌های نمایش پارامترها چه ارتباطی با فضای ذخیره‌سازی نظری آن‌ها دارد.</p>
