```python
# Q7
import numpy as np

students = np.array(
    [
        [2, 60, 45, 50],
        [5, 80, 70, 75],
        [1, 40, 35, 38],
        [8, 90, 88, 92],
        [4, 75, 60, 65],
        [7, 85, 80, 84],
        [3, 55, 50, 52],
        [6, 70, 72, 78],
    ]
)

x = students[:, :3]

x_min = np.min(x, axis=0)
x_max = np.max(x, axis=0)

new_student = np.array([5, 75, 68])

new_student_scaled = (new_student - x_min) / (x_max - x_min)
print(new_student_scaled)

x_scaled = (x - x_min) / (x_max - x_min)

distance = np.sqrt(np.sum((new_student_scaled - x_scaled) ** 2, axis=1))
# distance = np.linalg.norm(new_student_scaled - x_scaled, axis=1)
print(f"distance:\n{distance}")

print(f"Nearest student: {students[np.argmin(distance)]}")

final_score = students[:, -1]
level_label = np.where(
    final_score < 60, "weak", np.where(final_score < 80, "normal", "strong")
)

print(f"Nearest class label:\n{level_label[np.argmin(distance)]}")
```

# Question 8

```python
# import
import numpy as np

# variables
new_students = np.array([[2, 50, 40], [6, 82, 78], [4, 65, 58], [8, 88, 85]])
students = np.array(
    [
        [2, 60, 45, 50],
        [5, 80, 70, 75],
        [1, 40, 35, 38],
        [8, 90, 88, 92],
        [4, 75, 60, 65],
        [7, 85, 80, 84],
        [3, 55, 50, 52],
        [6, 70, 72, 78],
    ]
)

x = students[:, :3]

x_min = np.min(x, axis=0)
x_max = np.max(x, axis=0)
x_scaled = (x - x_min) / (x_max - x_min)
new_student_scaled = (new_students - x_min) / (x_max - x_min)

level_label = np.where(
    final_score < 60, "weak", np.where(final_score < 80, "normal", "strong")
)
print(f"shape x:{x_scaled.shape}")
print(f"shape new samples: {new_student_scaled.shape}")
# 4.1.3  1.8.3
distance = np.linalg.norm(
    x_scaled[np.newaxis, :, :] - new_student_scaled[:, np.newaxis, :], axis=2
)
nearest_neighbor = distance.argmin(axis=1)
print(f"distances: \n{distance}")
print(f"nearest neighbor:\n{nearest_neighbor}")
labels = level_label[nearest_neighbor]
for i, label in enumerate(labels, start=1):
    print(f"student {i}, label = {label}")
distance.shape
```

# Question 9

```python
# Q9

mapping = {"weak": "fail", "normal": "pass", "strong": "pass"}
predicted_labels = level_label[nearest_neighbor]
predicted_labels = np.vectorize(mapping.get)(predicted_labels)
print(predicted_labels)

true_labels = np.array(["fail", "pass", "pass", "pass"])

print(f"true_labels:{true_labels.shape}")
print(f"predicted_labels:{predicted_labels.shape}")

correct_predictions = true_labels == predicted_labels
print(f"correct_predictions:{correct_predictions}")

number_of_correct_predictions = np.sum(correct_predictions)
print(f"number_of_correct_predictions:{number_of_correct_predictions}")

accuracy = (number_of_correct_predictions / len(predicted_labels)) * 100

print(f"accuracy:{accuracy}")

# Accuracy ==> دقت
# Accuracy = 0.75 ==> از هر 4 پیش بینی 3 تا درست بوده است.
```

```python
"""This question will tell us somthing about regression"""

# imports
import numpy as np

# variables
actual_scores = np.array([50, 75, 38, 92, 65])
predicted_scores = np.array([55, 70, 45, 88, 60])
# regression error
error = predicted_scores - actual_scores
# error output
print(f"error of prediction vs actual values:\n{error}")
# absolute error
abs_error = np.abs(predicted_scores - actual_scores)
print(f"absolute error of prediction vs actual values:\n{abs_error}")

# MAE metric
mae_result = np.mean(abs_error)
# output
print(f"MAE metric:\n{mae_result}")
```
