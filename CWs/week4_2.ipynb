# Question 1

```python
##Q1 and Q2
import numpy as np

customers = np.array(
    [
        [22, 2, 15, 1],
        [25, 4, 20, 2],
        [47, 10, 120, 12],
        [52, 12, 150, 15],
        [35, 6, 60, 6],
        [38, 7, 70, 7],
        [19, 1, 10, 1],
        [60, 8, 130, 10],
        [41, 9, 90, 9],
        [28, 5, 35, 4],
        [33, 6, 55, 5],
        [55, 11, 160, 14],
    ]
)

x = customers

x_min = np.min(x, axis=0)
x_max = np.max(x, axis=0)
x_mean = np.mean(x, axis=0)
x_std = np.std(x, axis=0)
print(x_min)
print(x_max)
print(x_mean)
print(x_std)
hightest_mean = np.argmax(x_mean)
print(hightest_mean)
hightest_mstd = np.argmax(x_std)
print(hightest_mstd)
```

# Question 3

```python
# scaling dataset

X_scaled = (x - x_min) / (x_max - x_min)
print(f"scaled dataset:\n{X_scaled}")
x_scaled_mean = np.mean(X_scaled, axis=0)
print(f"mean before scaling:\n{x_mean}\nscaled dataset features mean:\n{x_scaled_mean}")
```

# Question 4

```python
# variables
ref_points = np.array(
    [
        # low activity
        [23, 2, 15, 1],
        # normal
        [32, 5, 45, 4],
        # loyal
        [42, 8, 85, 8],
        # VIP
        [55, 12, 150, 14],
    ]
)
# scaling new dataset
ref_points_scaled = (ref_points - x_min) / (x_max - x_min)
print(f"scaled new dataset:\n{ref_points_scaled}")
ref_names = np.array(["low activity", "normal", "loyal", "VIP"])

# distance
distances = np.linalg.norm(
    ref_points_scaled[:, np.newaxis, :] - X_scaled[np.newaxis, :, :], axis=2
).T
nearest_idx = distances.argmin(axis=1)

# mapping to labels
mapping = {0: "low activity", 1: "normal", 2: "loyal", 3: "VIP"}
predicted_labels = np.vectorize(mapping.get)(nearest_idx)
print(predicted_labels)
for i, (customer, label) in enumerate(zip(customers, predicted_labels), start=1):
    print(f"{i}. Customer {customer}\nwith label = {label}\n{"--"*10}")
```

# Question 7

```python
# Q7
customers = np.array(
    [
        [22, 2, 15, 1],
        [25, 4, 20, 2],
        [47, 10, 120, 12],
        [52, 12, 150, 15],
        [35, 6, 60, 6],
        [38, 7, 70, 7],
        [19, 1, 10, 1],
        [60, 8, 130, 10],
        [41, 9, 90, 9],
        [28, 5, 35, 4],
        [33, 6, 55, 5],
        [55, 11, 160, 14],
    ]
)

outlier_customer = np.array([[30, 1, 500, 1]])

customers_with_outlier = np.concatenate((customers, outlier_customer))

print(f"customers_with_outlier:\n{customers_with_outlier}")
print(f"customers_with_outlier shape is ==> {customers_with_outlier.shape}")


customers_with_outlier_min = np.min(customers_with_outlier, axis=0)
customers_with_outlier_max = np.max(customers_with_outlier, axis=0)

customers_with_outlier_scaled = (
    customers_with_outlier - customers_with_outlier_min
) / (customers_with_outlier_max - customers_with_outlier_min)

print(f"customers_with_outlier_scaled:\n{customers_with_outlier_scaled}")

center = np.mean(customers_with_outlier_scaled, axis=0)
print(f"center:{center}")

distance = np.linalg.norm(customers_with_outlier_scaled - center, axis=1)
print(f"distance:\n{distance}")

highest_distance = np.argmax(distance)
print(f"highest_distance: {highest_distance}")  # 12 ==> outlier

# 1 ==> outlier : داده پرت
# 2 ==>  مقدار avg_order_value بسیار بزرگی دارد، اما تعداد خرید او کم است
# 3 ==> خیر برچسب نداشتیم
# 4 ==> Unsupervised Learning
```
