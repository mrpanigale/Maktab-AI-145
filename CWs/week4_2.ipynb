```python
# imports
import numpy as np

# variables
customers = np.array(
    [
        [22, 2, 15, 1],
        [25, 4, 20, 2],
        [47, 10, 120, 12],
        [52, 12, 150, 15],
        [35, 6, 60, 6],
        [38, 7, 70, 7],
        [19, 1, 10, 1],
        [60, 8, 130, 10],
        [41, 9, 90, 9],
        [28, 5, 35, 4],
        [33, 6, 55, 5],
        [55, 11, 160, 14],
    ]
)
# results
print(
    "ndim ",
    customers.ndim,
    "shape",
    customers.shape,
    "size",
    customers.size,
    "dtype",
    customers.dtype,
    "\n",
    sep="\n",
)
```

# Question 2

```python
# imports
import numpy as np

# variables
x = customers
# statistics
x_min = np.min(x, axis=0)
x_max = np.max(x, axis=0)
x_mean = np.mean(x, axis=0)
x_std = np.std(x, axis=0)
# outputs
print(f"min :{x_min}")
print(f"max: {x_max}")
print(f"mean: {x_mean}")
print(f"std: {x_std}")

# the features with most mean and std
highest_mean = np.argmax(x_mean)
print(f"highest mean:\n{highest_mean}")
highest_std = np.argmax(x_std)
print(f"highest std:\n{highest_std}")
```

# Question 3

```python
# scaling dataset
# Min Max scaling
X_scaled = (x - x_min) / (x_max - x_min)
# outputs
print(f"scaled dataset:\n{X_scaled}")
x_scaled_mean = np.mean(X_scaled, axis=0)
print(
    f"means before scaling:\n{x_mean}\nscaled dataset features mean:\n{x_scaled_mean}"
)
```

```python
# imports
import numpy as np


# functions
def accuracy(ytrain, ypred):
    """this function return error and accuracy score"""
    corrects = np.sum(ytrain == ypred)
    wrongs = len(ytrain) - corrects
    acc = corrects / len(ytrain)
    error = wrongs / len(ytrain)
    return acc, error


# variables
y_train = np.array(["pass", "fail", "pass", "pass", "fail", "pass"])
y_test = np.array(["pass", "fail", "fail", "pass"])

# model A evaluate
model_A_train_pred = np.array(["pass", "fail", "pass", "pass", "fail", "pass"])
model_A_test_pred = np.array(["fail", "pass", "pass", "fail"])

# model B evaluate
model_B_train_pred = np.array(["pass", "pass", "pass", "pass", "pass", "pass"])
model_B_test_pred = np.array(["pass", "pass", "pass", "pass"])
# model C evaluate
model_C_train_pred = np.array(["pass", "fail", "pass", "fail", "fail", "pass"])
model_C_test_pred = np.array(["pass", "fail", "fail", "pass"])

# measure train acc for A,B,C
a_train_acc = accuracy(y_train, model_A_train_pred)[0]
b_train_acc = accuracy(y_train, model_B_train_pred)[0]
c_train_acc = accuracy(y_train, model_C_train_pred)[0]
# measure test acc for A,B,C
a_test_acc = accuracy(y_test, model_A_test_pred)[0]
b_test_acc = accuracy(y_test, model_B_test_pred)[0]
c_test_acc = accuracy(y_test, model_C_test_pred)[0]

print(f"Model A train accuracy: {a_train_acc:.2f}\n\
Model B train accuracy: {b_train_acc:.2f}\n\
Model C train accuracy: {c_train_acc:.2f}\n")

print(f"Model A test accuracy: {a_test_acc:.2f}\n\
Model B test accuracy: {b_test_acc:.2f}\n\
Model C test accuracy: {c_test_acc:.2f}\n")
# Model A => Best performance on train data
# Model C => Best performance on test data

# if we only check train accuracy there is risk of overfitting ( model loose performance on new datasets )
```

```python
# imports
import numpy as np


# function
def mae_score(ytrain, ypred):
    """This function measure MAE for regression models"""
    error = ytrain - ypred
    return np.mean(np.abs(error))


# variables
# ground truth
y_train_reg = np.array([50, 75, 38, 92, 65, 84])
y_test_reg = np.array([55, 80, 60, 90])
# models
# Model A
model_A_train_pred = np.array([50, 75, 38, 92, 65, 84])
model_A_test_pred = np.array([30, 100, 40, 120])
# Model B
model_B_train_pred = np.array([60, 60, 60, 60, 60, 60])
model_B_test_pred = np.array([60, 60, 60, 60])
# Model C
model_C_train_pred = np.array([52, 73, 40, 88, 66, 82])
model_C_test_pred = np.array([57, 78, 63, 88])

# MAE train for A,B,C
model_a_mae_train = mae_score(y_train_reg, model_A_train_pred)
model_b_mae_train = mae_score(y_train_reg, model_B_train_pred)
model_c_mae_train = mae_score(y_train_reg, model_C_train_pred)

print(f"Model A train MAE score: {model_a_mae_train:.2f}\n\
Model B train MAE score: {model_b_mae_train:.2f}\n\
Model C train MAE score: {model_c_mae_train:.2f}\n")

# MAE test for A,B,C
model_a_mae_test = mae_score(y_test_reg, model_A_test_pred)
model_b_mae_test = mae_score(y_test_reg, model_B_test_pred)
model_c_mae_test = mae_score(y_test_reg, model_C_test_pred)

print(f"Model A test MAE score: {model_a_mae_test:.2f}\n\
Model B test MAE score: {model_b_mae_test:.2f}\n\
Model C test MAE score: {model_c_mae_test:.2f}\n")

print("Model A overfitted\nModel B underfitted \nModel C good fitted")
```

```python
# imports
import numpy as np


# functions
def mae_score(ytrain, ypred, return_what="mae"):
    """This function measure MAE for regression models"""
    error = ytrain - ypred
    if return_what == "mae":
        return np.mean(np.abs(error))
    return np.abs(error)


# variables
# study_hours, attendance, previous_score
X = np.array(
    [
        [2, 60, 45],
        [5, 80, 70],
        [1, 40, 35],
        [8, 90, 88],
        [4, 75, 60],
        [7, 85, 80],
    ]
)
y = np.array([50, 75, 38, 92, 65, 84])

# simple model
prediction_simple = X[:, 0] * 10

# complex model
prediction_better = 0.4 * X[:, 0] * 10 + 0.3 * X[:, 1] + 0.3 * X[:, 2]

# abs error for each model
# simple
abs_simple = mae_score(y, prediction_simple, "abs")
# complex
abs_complex = mae_score(y, prediction_better, "abs")
# out put
print(f"simple model abs error:\n{abs_simple}\n\
complex model abs error:\n{abs_complex}")
# mae for each model
# simple
mae_simple = mae_score(y, prediction_simple)
# complex
mae_complex = mae_score(y, prediction_better)
# out put
print(f"simple model mae score:\n{mae_simple:.2f}\n\
complex model mae score:\n{mae_complex:.2f}")

# مدل ساده در اینجا به علت سادگی و عدم استفاده از فیچر های کافی پیچیدگی کافی باری حل مسئله رو نداشت
# ولی استفاده بیش از حد از فیچر هم مناسب نیست و ممکنه باعث بیش برازش یا اشباع شدن مدل بشه
# پس در ماشین لرنینگ هیچ چیز دائمی و همیشه مناسبی وجود ندارد و همه چیز به نیاز های مسئله بستگی دارد که در اینجا
# مدل ساده و تک فیچر نتوانست خوب عمل کند
```

```python
# imports
import numpy as np

# variables
models = np.array(["Model A", "Model B", "Model C"])
train_error = np.array([2, 25, 8])
test_error = np.array([40, 30, 10])

# measure gap
gap = train_error - test_error
print(train_error)
print(test_error)
print(gap)
# if train error be small and
#  test error be big the out put will be negative so argmin will help
case_1 = gap.argmin()
print(f"{models[case_1]} is overfitting")
# if we sum test error with gap the bigest out put
#  is the model with big error in train and test because
# if both error be big the output will be small
#  and we can't find out errors are big or small
case_2 = test_error + gap
print(f"{models[case_2.argmax()]} is Underfit model")

# in this case smallest mean gap is good fit
#  because both error are small and the mean will be small too.
case_3 = np.mean([train_error, test_error], axis=0)
print(f"{models[case_3.argmin()]} is good fit ")

# model with smallest test error
case_4 = np.argmin(test_error)
print(f"{models[case_4]} has best Generalization")
# because test datas are unseen datas and if model
# work nice with unseen datas , the model is great.
```
