```python
import requests

url = "http://127.0.0.1:11434/api/chat"

payload = {
    "model": "qwen2.5:0.5b",
    "messages":[
        {
            "role":"system",
            "content":"You are a concise teaching assistant. Answer in one English sentence."
        },
        {
            "role":"user",
            "content": "What is quantization?"
        }
    ],
    "stream": False,
    "options":{
        "temperature":0,
        "num_predict":40
    }
}

response = requests.post(url, json=payload)
response.raise_for_status()

data = response.json()

print(data["message"]["content"])
```

```python
import requests

url = "http://127.0.0.1:11434/v1/chat/completions"

payload = {
    "model": "qwen2.5:0.5b",
    "messages": [
        {
            "role": "system",
            "content": "You are a concise teaching assistant. Answer in one English sentence."
        },
        {
            "role": "user",
            "content": "What is quantization?"
        }
    ],
    "temperature": 0,
    "max_tokens": 40
}

response = requests.post(url, json=payload)
response.raise_for_status()

data = response.json()
print(data["choices"][0]["message"]["content"])
```
