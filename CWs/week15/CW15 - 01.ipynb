````python
def complete_gap(prompt,temperature=0,stop=None):

    response = client.chat.completions.create(model="gpt-5-mini",stop=stop,temperature=temperature,messages=[{"role": "user", "content":str(prompt)}])

    return response.choices[0].message.content
````

Q 3: فیو شات کمک میکنه هم فرمت رو یادبگیره ، هم پاسخ درست رو ! اگر استفاده نشه ممکنه لیبل های غلط داشته باشیم



````python
import math

def normalize_label(token):
    return token.strip().upper()


def extract_label_probs(choice, labels=("A", "B")):
    """
    Find the first generated position where all target labels
    appear in top_logprobs, then return normalized probabilities.
    """

    if choice.logprobs is None or not choice.logprobs.content:
        return None

    for token_info in choice.logprobs.content:

        top_logprobs = token_info.get("top_logprobs", [])

        raw_probs = {label: 0.0 for label in labels}
        seen = set()

        for item in top_logprobs:
            token = normalize_label(item["token"])
            logprob = item["logprob"]

            if token in raw_probs:
                raw_probs[token] = max(
                    raw_probs[token],
                    math.exp(logprob)
                )
                seen.add(token)

        if all(label in seen for label in labels):

            total = sum(raw_probs.values())

            if total == 0:
                return None

            return {
                label: raw_probs[label] / total
                for label in labels
            }

    return None
````
