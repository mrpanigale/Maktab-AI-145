## Q3

```python
# Q3
import numpy as np

np.random.seed(0)
questions = np.array(["Q1", "Q2", "Q3", "Q4", "Q5"])
answers = np.array(["B", "D", "A", "C", "B"])
shuffle_indexes = np.random.permutation(len(questions))
questions_shuffled = questions[shuffle_indexes]
answers_shuffled = answers[shuffle_indexes]
print(questions_shuffled)
print(answers_shuffled)

new_questions = np.array(["Q6", "Q7"])
new_answers = np.array(["A", "D"])

stacked_question = np.hstack((questions_shuffled, new_questions))

stacked_answers = np.hstack((answers_shuffled, new_answers))

print(f"stacked_question: {stacked_question}")
print(f"stacked_answers: {stacked_answers}")

difficulty = np.array([1, 2, 1, 3, 2, 2, 3])

result = np.vstack((stacked_question, stacked_answers, difficulty)).T
print(f"result: {result}")
print(f"result shape: {result.shape}")
```
