## کد اولیه
سلول زیر را اجرا کنید.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

n_samples = 240

X = rng.normal(size=(n_samples, 3))

true_w = np.array([
    1.6,
    -1.1,
    0.8
])

true_b = -0.2

true_logits = X @ true_w + true_b

true_prob = 1 / (1 + np.exp(-true_logits))

y = (
    rng.random(n_samples)
    < true_prob
).astype(float)

feature_names = [
    "waiting_time",
    "first_response_quality",
    "reopen_history"
]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Positive rate:", y.mean())
```

## توابع آماده‌ی شرکت

توابع زیر توسط تیم قبلی آماده شده‌اند و **نباید تغییر داده شوند**.

در این فعالیت قرار نیست فرمول داخلی Gradient را استخراج کنید. تمرکز روی **تفسیر Gradient و استفاده از آن در Gradient Descent** است.

</div>

```python
def sigmoid(z):
    z = np.clip(z, -50, 50)
    return 1 / (1 + np.exp(-z))


def dataset_bce(y, p):
    eps = 1e-12
    p = np.clip(p, eps, 1 - eps)

    loss = -np.mean(
        y * np.log(p)
        +
        (1 - y) * np.log(1 - p)
    )

    return loss


def company_training_service(X, y, w, b):
    z = X @ w + b
    p = sigmoid(z)

    loss = dataset_bce(y, p)

    error_signal = p - y

    grad_w = (
        X.T @ error_signal
    ) / len(X)

    grad_b = np.mean(error_signal)

    return {
        "probability": p,
        "loss": loss,
        "error_signal": error_signal,
        "grad_w": grad_w,
        "grad_b": grad_b
    }
```

تابع زیر را کامل کنید.

تابع باید یکی از سه پیام زیر را برگرداند:

- `"Prediction should move higher"`

برای حالت سوم می‌توانید شرط `abs(signal) < 0.1` را در نظر بگیرید.

</div>

```python
def interpret_signal(signal):
    if abs(signal) < 0.1:
        return "Prediction is very close to target"
    if signal < 0:
        return "Prediction should move higher"
    else:
        return "Prediction should move lower"

for i in range(8):
    print(
        "Ticket:",
        i,
        "Actual:",
        y[i],
        "Prediction:",
        report["probability"][i],
        "Signal:",
        report["error_signal"][i],
        "\n",
        interpret_signal(report["error_signal"][i]),
    )
```

## بخش ۴ ـ ساخت Training Snapshot

تابع زیر را کامل کنید تا اطلاعات زیر را برگرداند:

- `loss`
- `mean_probability`
- `mean_abs_signal`
- `gradient_norm`

</div>

```python
def training_snapshot(
    X,
    y,
    w,
    b
):

    result = company_training_service(
        X,
        y,
        w,
        b
    )

    grad_w = result["grad_w"]
    grad_b = result["grad_b"]

    gradient_norm = np.sqrt(
        np.sum(grad_w ** 2)
        +
        grad_b ** 2
    )

    return {
        "loss": result["loss"],
        "mean_probability": np.mean(result["probability"]),
        "mean_abs_signal": np.mean(np.abs(result["error_signal"])),
        "gradient_norm": gradient_norm
    }


print(training_snapshot(X, y, w, b))
```

## بخش ۱ ـ پیاده‌سازی Gradient Descent Update

تابع زیر را کامل کنید.

</div>

```python
def apply_gradient_step(
    w,
    b,
    grad_w,
    grad_b,
    learning_rate
):

    new_w = w - (grad_w * learning_rate)

    new_b = b - (grad_b * learning_rate)

    return new_w, new_b
```

## بخش ۲ ـ ساخت Training Loop

تابع زیر را کامل کنید.

در هر Step:

1. از `company_training_service` استفاده کنید.
2. Loss را ذخیره کنید.
3. Gradientها را دریافت کنید.
4. با `apply_gradient_step` پارامترها را Update کنید.

</div>

```python
def run_training(
    X,
    y,
    learning_rate,
    steps=50
):

    w = np.zeros(
        X.shape[1]
    )

    b = 0.0

    loss_history = []

    for step in range(steps):

        result = company_training_service(
            X,
            y,
            w,
            b
        )

        loss_history.append(
            result["loss"]
        )

        grad_w = result["grad_w"]
        grad_b = result["grad_b"]

        # TODO:
        # w و b را Update کنید.
        w ,b= apply_gradient_step(w,b,grad_w,grad_b,learning_rate)
    return (
        w,
        b,
        loss_history
    )

fitted_model = run_training(X,y, learning_rate)
print(f"W {fitted_model[0]},B {fitted_model[1]}\n Loss history {fitted_model[2]}")
```

## بخش ۱ ـ ساخت Training با Batchهای مختلف

تابع زیر را کامل کنید.

</div>

```python
def train_with_batches(
    X,
    y,
    batch_size,
    learning_rate=0.1,
    epochs=15,
    seed=42
):
    # random seed object to keep local effect of random seed
    rng = np.random.default_rng(
        seed
    )

    w = np.zeros(
        X.shape[1]
    )

    b = 0.0

    epoch_loss_history = []
    gradient_norm_history = []

    update_count = 0

    for epoch in range(epochs):

        indices = rng.permutation(
            len(X)
        )

        X_shuffled = X[indices]
        y_shuffled = y[indices]

        for start in range(
            0,
            len(X),
            batch_size
        ):

            end = (
                start
                +
                batch_size
            )

            X_batch = X_shuffled[start:end]
            y_batch = y_shuffled[start:end]

            result = company_training_service(
                X_batch,
                y_batch,
                w,
                b
            )

            grad_w = result["grad_w"]
            grad_b = result["grad_b"]

            gradient_norm = np.sqrt(
                np.sum(
                    grad_w ** 2
                )
                +
                grad_b ** 2
            )

            gradient_norm_history.append(
                gradient_norm
            )

            # TODO:
            # پارامترها را Update کنید.
            w , b = apply_gradient_step(w,b,grad_w,grad_b,learning_rate)

            update_count += 1

        full_result = company_training_service(
            X,
            y,
            w,
            b
        )

        epoch_loss_history.append(
            full_result["loss"]
        )

    return {
        "w": w,
        "b": b,
        "loss_history": epoch_loss_history,
        "gradient_norm_history": gradient_norm_history,
        "update_count": update_count
    }
```
