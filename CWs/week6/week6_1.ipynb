```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from sklearn.metrics import confusion_matrix

# ۱. تعریف متغیرها و توزیع‌ها (نمونه فرضی)
x = np.linspace(0, 20, 1000)
girl_dist = norm(loc=8, scale=2)
boy_dist = norm(loc=12, scale=3)

P_GIRL = 0.6
P_BOY = 0.4

# ۲. محاسبه چگالی‌های مشترک (Joint Probability Densities)
joint_girl = girl_dist.pdf(x) * P_GIRL
joint_boy = boy_dist.pdf(x) * P_BOY

# ۳. محاسبه احتمال کل (Evidence / Normalizer)
evidence = joint_girl + joint_boy

# ۴. محاسبه احتمال پسین نرمال‌شده (Posterior Probabilities)
posterior_girl = joint_girl / evidence
posterior_boy = joint_boy / evidence

# پیدا کردن نقطه‌ مرز تصمیم (جایی که احتمال هر دو کلاس برابر 0.5 می‌شود)
decision_boundary = x[np.argmin(np.abs(posterior_girl - posterior_boy))]

# ۵. رسم نمودار احتمال پسین
plt.figure(figsize=(8, 5))
plt.plot(x, posterior_girl, color="tab:red", linewidth=2, label=r"$P(\text{girl}|x)$")
plt.plot(x, posterior_boy, color="tab:blue", linewidth=2, label=r"$P(\text{boy}|x)$")

# رسم خط مرز تصمیم
plt.axvline(
    decision_boundary,
    color="black",
    linestyle="--",
    alpha=0.7,
    label=f"Decision Boundary ({decision_boundary:.2f})",
)

plt.xlabel("tav-hormone level")
plt.ylabel("Posterior Probability")
plt.title("Posterior Probabilities $P(\text{Class}|x)$")
plt.ylim(-0.05, 1.05)  # محدوده احتمال بین ۰ و ۱
plt.legend(loc="center right")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt

g_10 = np.random.normal(2, 4, 50)
class_10 = np.zeros(50)

g_20 = np.random.normal(8, 4, 50)
class_20 = np.ones(50)

x = np.hstack((g_10, g_20)).reshape(-1, 1)
y = np.hstack((class_10, class_20))


Xtrain, Xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, random_state=42)

plt.scatter(Xtrain, np.ones_like(ytrain), c=ytrain)
plt.title("train")
plt.show()
plt.scatter(Xtest, np.ones_like(ytest), c=ytest)
plt.title("test")
plt.show()

log_reg = LogisticRegression()
log_reg.fit(Xtrain, ytrain)
y_hat = log_reg.predict(Xtest)

confusion_matrix(ytest, y_hat)

disp = ConfusionMatrixDisplay.from_predictions(
    ytest, y_hat, display_labels=log_reg.classes_, cmap=plt.cm.Blues
)
plt.title("Confusion Matrix (from_predictions)")
plt.show()
print(f"accuracy {log_reg.score(Xtest,ytest)}")
print(f"accuracy train{log_reg.score(Xtrain,ytrain)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

np.random.seed(42)

n = 100
roast_time = np.random.uniform(8,18,n)
weight_loss = 2 + (0.9 * roast_time )+ np.random.normal(loc=0,scale=1.5,size=n) 



Xtrain,Xtest,ytrain,ytest = train_test_split(roast_time.reshape(-1,1),weight_loss,test_size = 0.3,random_state=42) 

linear_model = LinearRegression()
linear_model.fit(Xtrain,ytrain)
ypred = linear_model.predict(Xtest)

r2 = r2_score(ytest,ypred)
print(f"r2 score : {r2:.2f}")
print(f"MSE : {np.mean(np.sum((ytest - ypred)**2))}")
print(f"intercept : {linear_model.intercept_}")
print(f"coef : {linear_model.coef_[0]}")

plt.scatter(roast_time,weight_loss,label="data")
plt.plot(roast_time,linear_model.predict(roast_time.reshape(-1,1)),label = "model line",c='blue')
plt.legend()
plt.title("roast time vs weight loss")
plt.ylabel("weight loss")
plt.xlabel("roast time")
plt.grid()


np.random.seed(42)

n = 100
roast_time = np.random.uniform(8,18,n)
weight_loss = 2 + (0.9 * roast_time )+ np.random.normal(loc=0,scale=4,size=n) 



Xtrain,Xtest,ytrain,ytest = train_test_split(roast_time.reshape(-1,1),weight_loss,test_size = 0.3,random_state=42) 

linear_model = LinearRegression()
linear_model.fit(Xtrain,ytrain)
ypred = linear_model.predict(Xtest)

r2 = r2_score(ytest,ypred)
print(f"r2 score : {r2:.2f}")
print(f"MSE : {np.mean(np.sum((ytest - ypred)**2))}")
print(f"intercept : {linear_model.intercept_}")
print(f"coef : {linear_model.coef_[0]}")

plt.scatter(roast_time,weight_loss,label="more noised data")
plt.plot(roast_time,linear_model.predict(roast_time.reshape(-1,1)),label = "noisy model line",c='red')
plt.legend()
plt.title("roast time vs weight loss")
plt.ylabel("weight loss")
plt.xlabel("roast time")
plt.grid()



plt.show()

```
