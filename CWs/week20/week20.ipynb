### import

```python
from langchain_community.document_loaders import PyPDFLoader
import glob
from langchain_text_splitters import RecursiveCharacterTextSplitter
import re


pdf_paths = glob.glob("data/*")
```

### loader for each pdf



```python
loaders = [PyPDFLoader(path) for path in pdf_paths]

```

### read PDFs

```python
documents = []

for loader in loaders:
    docs = loader.load()
    documents.extend(docs)
```

```python
print(f"metadata: {documents[0].metadata.keys()}")

print(documents[0].page_content[:100])
```

```python
for i, doc in enumerate(documents):
    if "references" in doc.page_content.lower():
        print("Document index:", i)
        print("Page:", doc.metadata["page"])
        print(doc.page_content[:300])
        print("-" * 40)
```

```python

def is_reference_page(text):

    # بررسی عنوان References
    has_title = bool(re.search(
        r"^\s*(references|bibliography|works cited)\s*$",
        text,
        re.IGNORECASE | re.MULTILINE
    ))

    # بررسی الگوی شماره‌گذاری منابع مثل [1]، [2] و ...
    citations = re.findall(
        r"(?m)^\s*\[\d+\]",
        text
    )

    # بررسی تراکم ارجاعات
    return has_title and len(citations) >= 3
```

```python
filtered_documents = [
    doc for doc in documents
    if not is_reference_page(doc.page_content)
]
print(len(documents))
print(len(filtered_documents))
```
