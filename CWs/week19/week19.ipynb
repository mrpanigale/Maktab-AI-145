```python
import re

def tokens(text):
    return re.findall(r"\w+", text.casefold(), flags=re.UNICODE)
words = [" گزارش\u200cها", "گزارشها", "می\u200cشود", "میشود", " نمی\u200cشود ", "نمیشود"]
for word in words:
    print(f"✔️ { tokens(word)}")
```

```python
def prepare_for_search(text):
    return text
def find_docs(query):
    results = []
    q = tokens(prepare_for_search(query))
    for i,(k,v)  in enumerate(DOCS.items(),start=1):
        print()
        v = prepare_for_search(v)
        if set(q) & set(tokens(v)):
            results.append(k)
    if not results:
        results.append(f"no results matched with {query}")
    return results
queries =  [" گزارش\u200cها", "گزارشها", "می\u200cشود", "میشود", " نمی\u200cشود ", "نمیشود","نسخه"]
all_queries = []
for query in queries:
    this_turn = find_docs(query)
    all_queries.append(this_turn)
    print(query, this_turn)
```

> چون یک کلمه غیر تخصصی و رایج است که در هر سندی می تواند با نیم فاصله و بدون نیم فاصله وجود داشته باشد

> ممکن است هدف تحلیل احساسات کاربر یا موارد دیگر باشد و حفظ نیم فاصله حیاتی باشد


```python
def split_words(text, size, overlap):
    if not (
        isinstance(size, int)
        and isinstance(overlap, int)
        and 0 <= overlap < size
    ):
        raise ValueError("Invalid size or overlap")
    words = text.split()
    chunks = []
    start = 0
    while start < len(words):
        end = start + size
        if end > len(words):
            end = len(words)
        word = " ".join(words[start:end])
        chunks.append(word)
        if end == len(words):
            break
        start += size - overlap
    return chunks

assert split_words("a b c d e", 3, 0) == ["a b c", "d e"]
assert split_words("a b c d e", 3, 1) == ["a b c", "c d e"]
assert split_words("", 3, 1) == []
print("Tests passed!")
```

```python
def contains_phrase(chunks, phrase):
    phrase_nospace = phrase.replace(" ", "")
    for chunk in chunks:
        if phrase in chunk or phrase_nospace in chunk.replace(" ", ""):
            return True
    return False

assert contains_phrase(["hello world", "python test"], "helloworld") is True
assert contains_phrase(["hello", "world"], "hello world") is False

target_phrase = ("ضبط را متوقف کنید و باتری را به ایستگاه شارژ شماره دو ببرید")
# target_phrase = (
#     " ضبط را متوقف کنید و باتری را"
#     ".به ایستگاه شارژ شماره دو ببرید "
# )

for overlap in [0, 4, 8]:
    chunks = split_words(DOCS["باتری"], 20, overlap)
    print("\nOverlap:", overlap)
    print("Number of chunks:", len(chunks))
    print("Phrase found:", contains_phrase(chunks, target_phrase))
    for i, chunk in enumerate(chunks, 1):
        print(i, chunk)
```
