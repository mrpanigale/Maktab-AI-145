```python
#part 1
item_value_counts = df["ItemsCount"].value_counts()
print(item_value_counts)
Category_value_counts = df["Category"].value_counts()
print(Category_value_counts)
#part2
Electronics = df[df["Category"] == "Electronics"]["ItemsCount"].sum()
print(f"Electronics: {Electronics}")
Books = df[df["Category"] == "Books"]["ItemsCount"].sum()
print(f"Books: {Books}")
Clothing = df[df["Category"] == "Clothing"]["ItemsCount"].sum()
print(f"Clothing: {Clothing}")
Beauty = df[df["Category"] == "Beauty"]["ItemsCount"].sum()
print(f"Beauty: {Beauty}")
#part3
City_value_counts = df["City"].value_counts()
print(City_value_counts)
Tehran = df[df["City"] == "Tehran"]["ItemsCount"].sum()
print(f"Tehran: {Tehran}")
Shiraz = df[df["City"] == "Shiraz"]["ItemsCount"].sum()
print(f"Shiraz: {Shiraz}")
Tabriz = df[df["City"] == "Tabriz"]["ItemsCount"].sum()
print(f"Tabriz: {Tabriz}")
Isfahan = df[df["City"] == "Isfahan"]["ItemsCount"].sum()
print(f"Isfahan: {Isfahan}")
Rasht = df[df["City"] == "Rasht"]["ItemsCount"].sum()
print(f"Rasht: {Rasht}")
#part 4
PaymentMethod_value_counts = df["PaymentMethod"].value_counts()
print(PaymentMethod_value_counts)
#part5
Returned_value_counts = df["Returned"].value_counts()
print(Returned_value_counts)
```
