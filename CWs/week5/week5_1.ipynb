# Question 8 

```python
# measure OrderValu Quantiles 25 50 75 90 %
quan_25 = df["OrderValue"].quantile(0.25)
quan_50 = df["OrderValue"].quantile(0.50)
quan_75 = df["OrderValue"].quantile(0.75)
quan_90 = df["OrderValue"].quantile(0.90)

print(f"Quantile 25%: {quan_25}\nQuantile 50%: {quan_50}\nQuantile 75%: {quan_75}\nQuantile 90%: {quan_90}\n")
df["OrderValue"].median()

# Quantile 50% or Q2 == median
# 25 percents of datas are equal or less than 147.5
#75 percents of datas are equal or less than from Q3
#90% percentile tells 90 percents of datas are equal or less than Q 90%
# if an ordervalue is more than Q 90% the order considred as luxury order and 10 percent in the end of range
# if IQR sais it's outlier it is ,  large values are not always outlier

```

# Question 9 

```python
IQR = quan_75 - quan_25
lower = quan_25 - (1.5* IQR)
upper = quan_75 + (1.5* IQR)
mask = (df["OrderValue"] >= upper) | (df['OrderValue'] <= lower)
outliers = df[mask]
display(outliers)

# اینکه یک اوتلیر انومالی یا اشتباه باشد بستگی به مسئله دارد در اینجا قیمت سفارش بالا بوده و مشکلی وجود ندارد و داده اشتباه نیست 
```

```python
# scatter plot on DeliveryMinute , CustomerRating
corr = df["CustomerRating"].corr(df["DeliveryMinutes"])
plt.scatter(df["DeliveryMinutes"],df["CustomerRating"],c="blue")
plt.ylabel("Customer Rating")
plt.xlabel("Delivery Minutes")
plt.title(f"correlation coef = {corr:.2f}")
plt.grid()
plt.show()
```

# Question 11

```python
cus_del = df["CustomerRating"].corr(df["DeliveryMinutes"])
plt.scatter(df["DeliveryMinutes"],df["CustomerRating"],c="blue")
plt.ylabel("Customer Rating")
plt.xlabel("Delivery Minutes")
plt.title(f"correlation coef = {cus_del:.2f}")
plt.grid()
plt.show()

item_ord = df["ItemsCount"].corr(df["OrderValue"])
plt.scatter(df["ItemsCount"],df["OrderValue"],c="red")
plt.ylabel("OrderValue")
plt.xlabel("ItemsCount")
plt.title(f"correlation coef = {item_ord:.2f}")
plt.grid()
plt.show()

dis_ord = df["DiscountPercent"].corr(df["OrderValue"])
plt.scatter(df["DiscountPercent"],df["OrderValue"],c="green")
plt.xlabel("DiscountPercent")
plt.ylabel("OrderValues")
plt.title(f"correlation coef = {dis_ord:.2f}")
plt.grid()
plt.show()

del_ord = df["OrderValue"].corr(df["DeliveryMinutes"])
plt.scatter(df["OrderValue"],df["DeliveryMinutes"],c="yellow")
plt.xlabel("OrderValue")
plt.ylabel("DeliveryMinutes")
plt.title(f"correlation coef = {del_ord:.2f}")
plt.grid()
plt.show()
```
