## ⁦Task 3⁩: طراحی ⁦Schema⁩ (۲۵ دقیقه · ۱۵ امتیاز + ۵ امتیاز اضافه)

یک `schema` هم به مدل کمک می‌کند بفهمد چگونه از ⁦Tool⁩ استفاده کند و هم برای کد شما مشخص می‌کند چه ورودی‌هایی معتبرند. هرچه کار یک ⁦Tool⁩ محدودتر و مشخص‌تر باشد، ایمن‌سازی آن ساده‌تر است.

### ۳.۱ اصلاح ⁦Schema⁩ (۵ امتیاز)

یک کارآموز، این توضیح را برای ⁦Tool⁩ مورد استفادهٔ ⁦ORBIT⁩ نوشته است:

```json
{
  "type": "function",
  "function": {
    "name": "do_stuff",
    "description": "does things on the station",
    "parameters": {
      "type": "object",
      "properties": {
        "cmd": {"type": "string"},
        "n": {"type": "number"}
      }
    }
  }
}
```

- **(الف)** دست‌کم ۴ نقطه‌ضعف آن را نام ببرید. (۲ امتیاز)
- **(ب)** در خانهٔ کد بعدی، یک `schema` کامل و اصلاح‌شده برای `airlock_cycle(airlock_id, direction)` بنویسید. مقدار `airlock_id` باید عدد صحیح بین ۱ تا ۴ باشد. مقدار `direction` فقط می‌تواند `"in"` یا `"out"` باشد. هر دو آرگومان اجباری‌اند و فیلدهای اضافی باید رد شوند. متن `description` هم باید به مدل بگوید چه زمانی از این ⁦Tool⁩ استفاده کند، این ⁦Tool⁩ وضعیت واقعی را تغییر می‌دهد و اجرای آن نیازمند تأیید انسان است. (۳ امتیاز: ۱ امتیاز برای نوع داده، محدوده و مقادیر مجاز؛ ۱ امتیاز برای `required` و نپذیرفتن فیلد اضافه؛ ۱ امتیاز برای توضیح مناسب)

````python
import os
from getpass import getpass
from openai import OpenAI

base_url = (os.getenv('LLM_URL') or '').rstrip('/')
if not base_url:
    raise RuntimeError('Set LLM_URL to the proxy base path (ask your instructor).')
api_key = os.getenv('OPENROUTER_API_KEY') or getpass('OpenRouter API key (input is hidden): ').strip()
if not api_key:
    raise RuntimeError('No API key provided.')

OPENROUTER_MODEL = 'google/gemma-4-26b-a4b-it'
client = OpenAI(base_url=base_url, api_key=api_key, timeout=45.0, max_retries=0)
print('Client ready for', OPENROUTER_MODEL)
````

````python
# Live call: the model only PROPOSES. Nothing runs yet.
live_tools = [{
    'type': 'function',
    'function': {
        'name': 'oxygen_reserve',
        'description': 'Hours of oxygen left in a station module.',
        'parameters': {
            'type': 'object',
            'properties': {'module': {'type': 'string', 'enum': ['hab', 'lab', 'greenhouse', 'cargo']}},
            'required': ['module'],
            'additionalProperties': False,
        },
    },
}]

live_messages = [{'role': 'user', 'content': 'How long can the lab module keep breathing?'}]
live_response = client.chat.completions.create(
    model=OPENROUTER_MODEL, messages=live_messages, tools=live_tools,
)
live_msg = live_response.choices[0].message
print('content   :', live_msg.content)
print('tool_calls:', [c.model_dump() for c in (live_msg.tool_calls or [])])
````
