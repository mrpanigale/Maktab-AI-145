<h2>معرفی محصول</h2>
<p><em>Studio Linden</em> یک آرایشگاه است. مشتری‌ها با <strong>Luna</strong> گفتگو می‌کنند تا وقت‌های خالی را پیدا کنند، <strong>وقت رزرو کنند</strong>، <strong>زمان رزرو را تغییر دهند</strong> یا <strong>رزرو را لغو کنند</strong>. Luna یک <code>LLM</code> است، اما سالن اجازه نمی‌دهد مدل زبانی مستقیماً تقویم رزرو یا بخش مالی را تغییر دهد. هر کار واقعی را <strong>کنترلر Python شما</strong> انجام می‌دهد: پیشنهاد مدل را دریافت می‌کند، قوانین سالن را بررسی می‌کند، از مشتری تأیید می‌گیرد، تغییر را با اطمینان انجام می‌دهد و نمی‌گذارد Luna اطلاعاتی را به مشتری بگوید که سیستم تأیید نکرده است.</p>
<p>شما قرار است این کنترلر را با نام <code>SalonAssistant</code> بسازید.</p>

<h3>قوانین اصلی پروژه</h3>

<li>مدل فقط می‌تواند <strong>پیشنهاد</strong> بدهد. هر کاری که اطلاعات سالن را تغییر می‌دهد (<code>book</code>، <code>reschedule</code>، <code>cancel</code>) باید <strong>با کد خودتان</strong> و <strong>بعد از تأیید واضح مشتری در پیام بعدی</strong> اجرا شود.</li>

<h2>۴. قوانین سالن</h2>
<ul>
<li>سالن از <strong>دوشنبه تا شنبه، ساعت 09:00 تا 18:00</strong> باز است. شروع وقت‌ها در فاصله‌های ۳۰ دقیقه‌ای است و هر نوبت باید <strong>تا ساعت 18:00 تمام شود</strong>. آرایشگرها روزهای تعطیل و خدمات مشخصی دارند (اطلاعات <code>hw_world</code> را بخوانید).</li>
<li>هر مشتری می‌تواند حداکثر <strong>۳ رزرو فعال</strong> داشته باشد.</li>
<li><strong>هزینهٔ لغو رزرو</strong> بر اساس زمان باقی‌مانده تا شروع: <strong>۲۴ ساعت یا بیشتر ← رایگان</strong>؛ <strong>از ۲ ساعت تا کمتر از ۲۴ ساعت ← ۵۰٪ قیمت، با گردکردن رو به پایین</strong>؛ <strong>کمتر از ۲ ساعت ← ۱۰۰٪ قیمت</strong>؛ بعد از شروع نوبت، لغو ممکن نیست. دقیقاً ۲۴ ساعت مانده هم رایگان محسوب می‌شود.</li>
<li><strong>تغییر زمان رزرو</strong> با همان آرایشگر و همان خدمت فقط وقتی مجاز است که تا شروع نوبت <strong>حداقل ۲۴ ساعت</strong> باقی مانده باشد. زمان هر رزرو حداکثر <strong>یک بار</strong> تغییر می‌کند و وقت جدید باید خالی باشد.</li>
<li>همهٔ مبلغ‌ها در سیستم به شکل <strong>عدد صحیح برحسب سنت</strong> هستند؛ در پیام مشتری به یورو نمایش داده می‌شوند، مثل <code>€35.00</code>.</li>
</ul>
<h2>۵. قوانین کنترلر (هر سناریوی مخفی یکی از این موارد را بررسی می‌کند)</h2>
<ul>
<li>
<p><strong>R1 — هویت:</strong> هویت مشتری از <code>session</code> گرفته می‌شود، نه از مدل. رزرو مشتری دیگر باید <code>not_found</code> برگرداند؛ دقیقاً مثل رزروی که وجود ندارد.</p>
</li>
<li>
<p><strong>R2 — تأیید مشتری:</strong> فراخوانی ابزار تغییر‌دهنده از سوی مدل فقط یک <strong>پیشنهاد</strong> است. ورودی را بررسی کنید و قوانین را <strong>در همان لحظه</strong> بسنجید: سقف رزرو، تداخل زمانی، تطابق آرایشگر و خدمت، خالی‌بودن وقت، هزینهٔ لغو و قانون ۲۴ ساعت. اگر کار مجاز بود، آن را به‌عنوان <strong>اقدام در انتظار تأیید</strong> (<code>pending action</code>) در وضعیت <code>session</code> ذخیره کنید و <code>needs_confirmation</code> را با خلاصه‌ای برگردانید که <strong>خودتان از داده‌های سرور می‌سازید</strong>. سپس مدل از مشتری می‌پرسد. در <strong>پیام بعدی مشتری</strong>، تابع <code>interpret_reply</code> را اجرا کنید:</p>
<ul>
<li><code>'yes'</code> ← <strong>کد شما</strong> اقدام در انتظار را اجرا می‌کند؛ مدل نباید دوباره همان اقدام را پیشنهاد دهد.</li>
<li>هر پاسخ دیگری ← اقدام در انتظار حذف می‌شود. <strong>هر اقدام در انتظار فقط تا یک پیام بعدی مشتری اعتبار دارد.</strong></li>
</ul>
<p>تابع <code>interpret_reply</code> باید محتاط باشد: فقط موافقت روشن و بدون شرط، <code>'yes'</code> است؛ فقط مخالفت واضح، <code>'no'</code> است. پیامی که <strong>اطلاعات تازه</strong> دارد (مثل «بله ولی ساعت ۳» یا «بله و اصلاح ریش هم اضافه کن»)، سؤال می‌پرسد، زمان دارد یا بیش از چند کلمه است، باید <code>'unclear'</code> شود. در صورت شک، <code>'unclear'</code> برگردانید؛ <strong>هیچ‌وقت با حدس رزرو نکنید.</strong> در هر لحظه <strong>حداکثر یک اقدام در انتظار</strong> مجاز است؛ پیشنهاد تغییر دوم در همان مرحله باید <code>blocked</code> با خطای <code>pending_exists</code> برگرداند.</p>
</li>
<li>
<p><strong>R3 — پروتکل اجرا:</strong> پس از تأیید اقدام در انتظار، یک پیام <code>assistant</code> با یک فراخوانی ابزار ساختگی و <code>id</code> یکتا و سپس پیام <code>tool</code> متناظر با نتیجه را به تاریخچه اضافه کنید. بعد مدل را <strong>یک بار</strong> برای نوشتن پاسخ فراخوانی کنید. مدل می‌تواند مثل سایر مراحل فراخوانی ابزار دیگری هم انجام دهد. تاریخچه‌ای که به مدل می‌دهید باید همیشه <strong>معتبر</strong> باشد: هر پیام <code>tool</code> به یک فراخوانی پیشین از <code>assistant</code> پاسخ دهد و <code>id</code>ها تکراری نباشند.</p>
</li>
<li>
<p><strong>R4 — بررسی پیش از درخواست تأیید:</strong> هیچ‌وقت برای کاری که امکان‌پذیر نیست (سقف رزرو، تداخل، وقت اشغال‌شده، دیرشدن و مانند آن) تأیید نخواهید. به‌جایش <code>blocked</code> برگردانید و هرجا جدول مشخص کرده، گزینه‌های جایگزین ارائه دهید.</p>
</li>
<li>
<p><strong>R5 — اجرای بدون تکرار (<code>Idempotency</code>):</strong> مقدار <code>request_key</code> موردنیاز <code>backend</code> را <strong>خودتان</strong> بسازید؛ مثلاً از ترکیب <code>session</code>، شمارهٔ مرحلهٔ گفتگو و شمارهٔ ترتیبی. هر اقدام تأییدشده، حتی هنگام خطا، نباید باعث ایجاد دو رزرو شود.</p>
</li>
<li>
<p><strong>R6 — مدیریت خطا:</strong> برای عملیات <strong>خواندن</strong>، در صورت <code>BackendTimeout</code> دقیقاً <strong>یک بار دیگر</strong> تلاش کنید؛ تعداد تلاش‌ها در <code>audit.attempts</code> ثبت می‌شود. اگر دوباره خطا رخ داد، <code>error</code> / <code>timeout</code> برگردانید. برای عملیات <strong>تغییر</strong>، <code>BackendTimeout</code> ممکن است بعد از انجام تغییر یا قبل از آن رخ داده باشد. با <code>find_by_key</code> از <code>backend</code> وضعیت را بررسی کنید؛ اگر تغییر ثبت شده بود، موفقیت را گزارش کنید. اگر ثبت نشده بود، <strong>یک بار با همان کلید</strong> دوباره تلاش کنید. اگر آن بار هم <code>timeout</code> شد، نتیجه <code>error</code> / <code>unknown_outcome</code> است و <strong>نباید ادعای موفقیت کنید</strong>. اگر در یک مرحلهٔ گفتگو <strong>دو خطا یا بیشتر</strong> از نوع <code>timeout</code>، <code>unknown_outcome</code> یا خطای غیرمنتظره رخ داد، فقط <strong>یک</strong> درخواست ارجاع به پشتیبان با علت <code>tool_failures</code> بسازید. در تداخل ناشی از اقدام هم‌زمان (<code>SlotTaken</code>) باید <code>error</code> / <code>slot_taken</code> همراه گزینه‌های جایگزین برگردانید. اگر هنگام تصمیم‌گیری مشتری هزینهٔ لغو تغییر کرد (<code>fee_mismatch</code> از <code>backend</code>)، رزرو را لغو نکنید.</p>
</li>
<li>
<p><strong>R7 — محدودیت فراخوانی‌ها:</strong> در هر پیام مشتری حداکثر <strong>۶ فراخوانی مدل و ۸ فراخوانی ابزار</strong> مجاز است. مجموعهٔ فراخوانی‌های نامعتبر (خالی، غیر از فهرست <code>dict</code>ها، <code>id</code> گم‌شده یا تکراری حتی نسبت به مراحل قبلی همان <code>session</code>، ساختار اشتباه یا عبور از محدودیت) را <strong>اصلاً اجرا نکنید</strong>؛ پاسخ مشتری <code>SAFE_FALLBACK</code> است. اگر تعداد فراخوانی مدل هم به سقف برسد، <code>SAFE_FALLBACK</code> برگردانید.</p>
</li>
<li>
<p><strong>R8 — پاسخ مبتنی بر شواهد (<code>Grounding</code>):</strong> قبل از نشان‌دادن متن نهایی مدل، آن را بررسی کنید. هر زمان با قالب <code>HH:MM</code>، هر مبلغ یورو مانند <code>€10</code> یا <code>€10.00</code> و هر شناسهٔ رزرو مثل <code>B-n</code> باید در <strong>شواهد</strong> وجود داشته باشد؛ یعنی در خروجی ابزارهای همین <code>session</code> یا متن نوشته‌شده توسط مشتری. اگر پاسخ می‌گوید رزروی <strong>ثبت، لغو یا جابه‌جا شده</strong>، باید در <strong>همین مرحلهٔ گفتگو</strong> یک عملیات تغییر موفق وجود داشته باشد. در غیر این صورت <code>SAFE_FALLBACK</code> را برگردانید.</p>
</li>
<li>
<p><strong>R9 — مدیریت تاریخچه:</strong> پیام <code>system</code> باید خط <code>Now: YYYY-MM-DD HH:MM (&lt;Weekday&gt;)</code> را داشته باشد که از <code>clock()</code> ساخته شده است. هر بار حداکثر <strong>۲۴ پیام</strong> به مدل بفرستید. کوتاه‌کردن تاریخچه نباید هیچ پیام <code>tool</code> را بدون فراخوانی متناظر رها کند. اقدام در انتظار و شواهد باید محفوظ بمانند، چون جزو وضعیت برنامه‌اند نه صرفاً متن چت.</p>
</li>
<li>
<p><strong>R10 — ثبت رویدادها (<code>Audit</code>):</strong> در <code>assistant.audit</code> به‌ازای هر فراخوانی ابزار (از جمله اجرای اقدام تأییدشده) <strong>و</strong> هر پیام مشتری (با <code>tool</code> برابر <code>'(user)'</code>) یک رویداد ثبت کنید. هر رویداد باید <strong>دقیقاً</strong> این کلیدها را داشته باشد: <code>session, turn, call_id, tool, status, error_type, attempts, preview</code>. مقدار <code>preview</code> از <code>data</code> مربوط به <code>observation</code> یا متن مشتری ساخته می‌شود؛ ابتدا <strong>ایمیل و شمارهٔ تلفن را مخفی کنید</strong> و سپس متن را به ۸۰ کاراکتر کوتاه کنید. تاریخ‌هایی مثل <code>2026-10-20 09:00</code> نباید در این فرایند حذف شوند. هیچ‌وقت آرگومان‌های خام را ثبت نکنید.</p>
</li>
<li>
<p><strong>R11 — ارجاع به نیروی انسانی:</strong> ابزار <code>handoff_to_human</code> یا قانون <code>R6</code> با <code>backend.create_ticket({...})</code> یک درخواست پشتیبانی می‌سازد. داده‌ها باید شامل <code>customer_id</code>، <code>reason</code> و یک <code>summary</code> حداکثر ۳۰۰ کاراکتری باشند که <strong>شمارهٔ تلفن یا ایمیل نداشته باشد</strong>.</p>

</li>
</ul>

```python
# GIVEN — do not edit this cell (the grader uses its own copy)
import json, re, copy, time, datetime as dt
from types import SimpleNamespace

# The salon's clock is INJECTED. Never call datetime.now() in your assistant: use the `clock` you are given.
NOW = dt.datetime(2026, 10, 19, 10, 0)            # Monday 19 October 2026, 10:00

def fmt(t):      return t.strftime('%Y-%m-%d %H:%M')
def parse_dt(s): return dt.datetime.strptime(s, '%Y-%m-%d %H:%M')

SERVICES = {   # minutes are a multiple of 30; prices are INTEGER CENTS
    'cut':        {'name': 'Haircut',      'minutes': 30,  'price_cents': 3500},
    'beard':      {'name': 'Beard trim',   'minutes': 30,  'price_cents': 2000},
    'wash_style': {'name': 'Wash & style', 'minutes': 60,  'price_cents': 4500},
    'color':      {'name': 'Colour',       'minutes': 120, 'price_cents': 9000},
    'highlights': {'name': 'Highlights',   'minutes': 150, 'price_cents': 13000},
}
STYLISTS = {   # weekday(): Monday=0 ... Sunday=6
    'mara': {'name': 'Mara', 'services': {'cut', 'wash_style', 'color', 'highlights'}, 'days_off': {6}},
    'jon':  {'name': 'Jon',  'services': {'cut', 'beard', 'wash_style'},               'days_off': {2, 6}},
    'lea':  {'name': 'Lea',  'services': {'cut', 'color', 'highlights'},               'days_off': {0, 6}},
}
OPEN_H, CLOSE_H, SLOT_MIN, MAX_ACTIVE = 9, 18, 30, 3
SEED = [   # (customer, service, stylist, start)
    ('c_ada', 'cut',        'mara', '2026-10-20 10:00'),   # B-1  exactly 24h from NOW
    ('c_ada', 'color',      'lea',  '2026-10-23 14:00'),   # B-2
    ('c_ben', 'beard',      'jon',  '2026-10-19 15:00'),   # B-3  today, 5h away
    ('c_cy',  'cut',        'jon',  '2026-10-20 13:00'),   # B-4
    ('c_cy',  'wash_style', 'mara', '2026-10-22 11:00'),   # B-5
    ('c_cy',  'beard',      'jon',  '2026-10-24 16:00'),   # B-6  (c_cy has the maximum of 3 active bookings)
]

class BackendTimeout(Exception):
    """The salon system did not answer in time (the request may or may not have been applied for writes)."""
class SlotTaken(Exception):
    """Somebody else holds that slot."""
class BackendError(Exception):
    """An expected business refusal; .code is a short machine-readable reason."""
    def __init__(self, code):
        super().__init__(code); self.code = code

class SalonBackend:
    """The salon's booking system. READ it, do not edit it. Every public method is a possible failure point."""
    def __init__(self, now=NOW):
        self.now = now
        self.bookings, self.by_key, self.tickets = {}, {}, []
        self.next_id = 1
        self.call_log, self.write_log = [], []          # every method called / every COMMITTED write
        self.faults = {'fail_reads': 0, 'fail_next_write': 0, 'lose_next_write_response': 0}
        for customer, service, stylist, start in SEED:
            self._insert(customer, service, stylist, start)

    # ---------- helpers ----------
    def _log(self, name): self.call_log.append(name)
    def _read_fault(self):
        if self.faults['fail_reads'] > 0:
            self.faults['fail_reads'] -= 1
            raise BackendTimeout('read timeout')
    def _insert(self, customer, service, stylist, start):
        bid = f'B-{self.next_id}'; self.next_id += 1
        self.bookings[bid] = {'booking_id': bid, 'customer_id': customer, 'service_id': service, 'stylist_id': stylist,
                              'start': start, 'status': 'active', 'reschedules': 0, 'fee_cents': 0}
        return self.bookings[bid]
    def _busy(self, stylist, ignore=None):
        out = []
        for b in self.bookings.values():
            if b['status'] == 'active' and b['stylist_id'] == stylist and b['booking_id'] != ignore:
                s = parse_dt(b['start']); out.append((s, s + dt.timedelta(minutes=SERVICES[b['service_id']]['minutes'])))
        return out
    def _is_free(self, service, stylist, start, ignore=None):
        s = parse_dt(start); e = s + dt.timedelta(minutes=SERVICES[service]['minutes'])
        if s <= self.now or s.weekday() in STYLISTS[stylist]['days_off'] or service not in STYLISTS[stylist]['services']:
            return False
        if s.hour < OPEN_H or e > s.replace(hour=CLOSE_H, minute=0) or s.minute % SLOT_MIN:
            return False
        return all(e <= bs or s >= be for bs, be in self._busy(stylist, ignore))
    def _fee(self, booking):
        price = SERVICES[booking['service_id']]['price_cents']
        delta = parse_dt(booking['start']) - self.now
        if delta <= dt.timedelta(0): raise BackendError('already_started')
        if delta >= dt.timedelta(hours=24): return 0
        if delta >= dt.timedelta(hours=2): return price // 2
        return price
    def _replay(self, key, sig):
        entry = self.by_key.get(key)
        if entry is None: return None
        if entry['sig'] != sig: raise BackendError('key_reuse')
        return copy.deepcopy(entry['result'])
    def _write_faults(self):
        if self.faults['fail_next_write'] > 0:
            self.faults['fail_next_write'] -= 1
            raise BackendTimeout('write timeout (nothing was applied)')
    def _lost_response(self):
        if self.faults['lose_next_write_response'] > 0:
            self.faults['lose_next_write_response'] -= 1
            raise BackendTimeout('response lost (the write WAS applied)')

    # ---------- reads ----------
    def list_services(self):
        self._log('list_services'); self._read_fault(); return copy.deepcopy(SERVICES)
    def free_slots(self, service_id, date, stylist_id=None):
        self._log('free_slots'); self._read_fault()
        if service_id not in SERVICES: raise ValueError('unknown service')
        day = dt.datetime.strptime(date, '%Y-%m-%d')
        out = []
        for stylist in sorted(STYLISTS if stylist_id is None else [stylist_id]):
            t = day.replace(hour=OPEN_H)
            while t.hour < CLOSE_H:
                if self._is_free(service_id, stylist, fmt(t)):
                    out.append({'stylist_id': stylist, 'start': fmt(t)})
                t += dt.timedelta(minutes=SLOT_MIN)
        return sorted(out, key=lambda r: (r['start'], r['stylist_id']))
    def bookings_for(self, customer_id):
        self._log('bookings_for'); self._read_fault()
        return copy.deepcopy([b for b in self.bookings.values() if b['customer_id'] == customer_id])
    def find_by_key(self, request_key):
        """Look up the stored result of an earlier write. Never faults."""
        self._log('find_by_key')
        entry = self.by_key.get(request_key)
        return copy.deepcopy(entry['result']) if entry else None

    # ---------- writes (idempotent on request_key) ----------
    def book(self, customer_id, service_id, stylist_id, start, request_key):
        self._log('book'); self._write_faults()
        sig = ('book', customer_id, service_id, stylist_id, start)
        done = self._replay(request_key, sig)
        if done is None:
            if service_id not in SERVICES or stylist_id not in STYLISTS: raise BackendError('unknown_item')
            if sum(1 for b in self.bookings.values() if b['customer_id'] == customer_id and b['status'] == 'active') >= MAX_ACTIVE:
                raise BackendError('limit_reached')
            if not self._is_free(service_id, stylist_id, start): raise SlotTaken(start)
            done = copy.deepcopy(self._insert(customer_id, service_id, stylist_id, start))
            self.by_key[request_key] = {'sig': sig, 'result': done}
            self.write_log.append(('book', done['booking_id']))
        self._lost_response()
        return done
    def cancel(self, customer_id, booking_id, fee_cents, request_key):
        self._log('cancel'); self._write_faults()
        sig = ('cancel', customer_id, booking_id, fee_cents)
        done = self._replay(request_key, sig)
        if done is None:
            b = self.bookings.get(booking_id)
            if b is None or b['customer_id'] != customer_id: raise BackendError('not_found')
            if b['status'] != 'active': raise BackendError('already_cancelled')
            if fee_cents != self._fee(b): raise BackendError('fee_mismatch')
            b['status'], b['fee_cents'] = 'cancelled', fee_cents
            done = {'booking_id': booking_id, 'fee_cents': fee_cents, 'status': 'cancelled'}
            self.by_key[request_key] = {'sig': sig, 'result': done}
            self.write_log.append(('cancel', booking_id))
        self._lost_response()
        return copy.deepcopy(done)
    def reschedule(self, customer_id, booking_id, new_start, request_key):
        self._log('reschedule'); self._write_faults()
        sig = ('reschedule', customer_id, booking_id, new_start)
        done = self._replay(request_key, sig)
        if done is None:
            b = self.bookings.get(booking_id)
            if b is None or b['customer_id'] != customer_id: raise BackendError('not_found')
            if b['status'] != 'active': raise BackendError('already_cancelled')
            if parse_dt(b['start']) - self.now < dt.timedelta(hours=24): raise BackendError('too_late')
            if b['reschedules'] >= 1: raise BackendError('already_rescheduled')
            if not self._is_free(b['service_id'], b['stylist_id'], new_start, ignore=booking_id): raise SlotTaken(new_start)
            b['start'], b['reschedules'] = new_start, b['reschedules'] + 1
            done = copy.deepcopy(b)
            self.by_key[request_key] = {'sig': sig, 'result': done}
            self.write_log.append(('reschedule', booking_id))
        self._lost_response()
        return done
    def create_ticket(self, ticket):
        """Hand a case to a human. `ticket` must be a dict with customer_id, reason, summary."""
        self._log('create_ticket')
        ticket = {**ticket, 'ticket_id': f'T-{len(self.tickets) + 1}'}
        self.tickets.append(ticket)
        return ticket['ticket_id']

# every seed booking must be valid: build one backend and check the invariants once
_b = SalonBackend()
assert all(_b._is_free(b['service_id'], b['stylist_id'], b['start'], ignore=b['booking_id']) for b in _b.bookings.values()), 'seed data must be valid'
print('Salon backend ready:', len(_b.bookings), 'seed bookings. NOW =', fmt(NOW), '(Monday)')
```

```python
# GIVEN — do not edit this cell (the grader uses its own copy)
SAFE_FALLBACK = "I can't confirm that from the salon system. Let me check again, or I can connect you with a team member."

def call(cid, name, **arguments):      return {'id': cid, 'name': name, 'arguments': arguments}
def calls(*cs):                        return {'tool_calls': list(cs)}
def final(text):                       return {'final': text}

def check_transcript(messages):
    """Rules every model call must satisfy. Returns a list of violations (empty = a valid conversation)."""
    bad, open_ids, seen_ids = [], None, set()
    if not messages or messages[0].get('role') != 'system':
        bad.append('the first message must be the system prompt')
    for i, m in enumerate(messages):
        role = m.get('role')
        if role not in ('system', 'user', 'assistant', 'tool'):
            bad.append(f'message {i}: unknown role {role!r}'); continue
        if role == 'tool':
            if open_ids is None or m.get('tool_call_id') not in open_ids:
                bad.append(f'message {i}: tool message {m.get("tool_call_id")!r} does not answer a pending tool call')
            else:
                open_ids.discard(m['tool_call_id'])
            try: json.loads(m.get('content'))
            except Exception: bad.append(f'message {i}: tool content must be a JSON string')
            continue
        if open_ids:
            bad.append(f'message {i}: tool calls {sorted(open_ids)} were never answered')
        open_ids = None
        if role == 'assistant' and m.get('tool_calls'):
            ids = [c.get('id') for c in m['tool_calls']]
            if len(set(ids)) != len(ids) or seen_ids & set(ids):
                bad.append(f'message {i}: duplicate tool call id')
            seen_ids |= set(ids); open_ids = set(ids)
    if open_ids:
        bad.append(f'tool calls {sorted(open_ids)} were never answered')
    if len(messages) > 24:
        bad.append(f'{len(messages)} messages sent to the model (limit 24)')
    return bad

class TurnScriptedModel:
    """A scripted 'model'. Each scenario turn loads the steps the model will take; every model call consumes one step."""
    def __init__(self):
        self.queue, self.violations, self.max_len, self.n_calls = [], [], 0, 0
        self.tools_seen, self.snapshots, self.turn_label = None, [], ''
    def load(self, steps, label=''):
        self.queue, self.turn_label = list(steps), label
    def __call__(self, messages, tools=None):
        self.n_calls += 1
        self.tools_seen = tools
        self.max_len = max(self.max_len, len(messages))
        self.snapshots.append(copy.deepcopy(messages))
        for v in check_transcript(messages):
            self.violations.append(f'[{self.turn_label}] {v}')
        if not self.queue:
            self.violations.append(f'[{self.turn_label}] the assistant called the model more often than this scenario allows')
            return {'final': 'SCRIPT-EXHAUSTED'}
        step = self.queue.pop(0)
        return copy.deepcopy(step(messages) if callable(step) else step)
    def observations(self):
        """Every tool observation the model has been shown so far (parsed, in order of first appearance)."""
        seen, out = set(), []
        for snap in self.snapshots:
            for m in snap:
                if m.get('role') == 'tool' and m['tool_call_id'] not in seen:
                    seen.add(m['tool_call_id'])
                    try: out.append({**json.loads(m['content']), '_id': m['tool_call_id']})
                    except Exception: pass
        return out

def T(user, *steps, session='s1', before=None, after=None, has=(), lacks=(), equals=None):
    return dict(user=user, model=list(steps), session=session, before=before, after=after,
                has=list(has), lacks=list(lacks), equals=equals)

def active(backend, customer=None):
    return [b for b in backend.bookings.values() if b['status'] == 'active' and (customer is None or b['customer_id'] == customer)]
def new_bookings(backend):
    return [b for b in backend.bookings.values() if int(b['booking_id'][2:]) > len(SEED)]
def tool_events(assistant):
    return [e for e in assistant.audit if e.get('tool') != '(user)']

def run_scenario(sc, assistant_cls, trace=False):
    """Run one scenario against a fresh salon. Returns None (pass) or a short failure message."""
    backend = SalonBackend()
    if sc.get('setup'): sc['setup'](backend)
    model = TurnScriptedModel()
    try:
        assistant = assistant_cls(model, backend, lambda: backend.now)
        sessions = sc.get('sessions') or {'s1': sc.get('customer', 'c_ada')}
        for sid, cust in sessions.items():
            assistant.open_session(sid, cust)
        ctx = SimpleNamespace(backend=backend, assistant=assistant, model=model, replies=[], sessions=sessions)
        for n, turn in enumerate(sc.get('turns', []), 1):
            label = f'turn {n} ({turn["user"][:28]!r})'
            if turn['before']: turn['before'](backend)
            model.load(turn['model'], label)
            reply = assistant.handle(turn['session'], turn['user'])
            ctx.replies.append(reply)
            if trace:
                print(f'\n[{label}]\n  USER : {turn["user"]}\n  REPLY: {reply}')
            if not isinstance(reply, str): return f'{label}: handle() must return a string'
            if model.queue and not sc.get('loose'): return f'{label}: the assistant called the model fewer times than expected ({len(model.queue)} script steps unused)'
            for needle in turn['has']:
                if needle.lower() not in reply.lower(): return f'{label}: reply should contain {needle!r}; got {reply[:90]!r}'
            for needle in turn['lacks']:
                if needle.lower() in reply.lower(): return f'{label}: reply must not contain {needle!r}'
            if turn['equals'] is not None and reply != turn['equals']: return f'{label}: expected exactly {turn["equals"][:50]!r}; got {reply[:70]!r}'
            if turn['after']:
                try: turn['after'](ctx)
                except AssertionError as exc: return f'{label}: {exc}'
        if sc.get('check'):
            sc['check'](ctx)
        if model.violations: return model.violations[0]
        return None
    except AssertionError as exc:
        return f'{exc}' or 'assertion failed'
    except NotImplementedError as exc:
        return f'not implemented yet ({exc})'
    except Exception as exc:
        return f'{type(exc).__name__}: {str(exc)[:110]}'

def run_suite(scenarios, assistant_cls, title='Public suite', total_points=None, verbose=True):
    results, earned, wsum = [], 0.0, sum(s['weight'] for s in scenarios)
    for sc in scenarios:
        err = run_scenario(sc, assistant_cls)
        results.append((sc, err))
        pts = (total_points or wsum) * sc['weight'] / wsum
        earned += 0 if err else pts
        if verbose:
            print(('✓' if err is None else '✗'), f"{sc['id']} {sc['title']}", '' if err is None else f'→ {err}')
    passed = sum(1 for _, e in results if e is None)
    print(f'{title}: {passed}/{len(scenarios)} scenarios passed' + (f'  →  {earned:.1f}/{total_points} points' if total_points else ''))
    return earned

def trace(scenario_id, assistant_cls, scenarios=None):
    """Debug helper: run ONE scenario and print every user message and reply, the failure (if any), and the audit trail."""
    pool = scenarios if scenarios is not None else PUBLIC
    sc = next(s for s in pool if s['id'] == scenario_id)
    print(f"=== {sc['id']} {sc['title']} ===")
    err = run_scenario(sc, assistant_cls, trace=True)
    print('\nRESULT:', 'PASS' if err is None else f'FAIL → {err}')
```

```python
SAFE_FALLBACK  # provided above

def interpret_reply(text):
    raw = text.lower().strip()

    # پاسخ دارای ساعت یا عدد، تأیید ساده و بی‌قیدوشرط نیست.
    if re.search(r"\d", raw):
        return "unclear"

    normalized = re.sub(r"[^a-z'\s]", "", raw).strip()
    words = re.findall(r"[a-z']+", normalized)

    if "yes" in words and "no" in words:
        return "unclear"

    yes_replies = {
        "yes", "y", "ok", "okay", "sure", "absolutely",
        "go ahead", "that works", "sounds good", "perfect",
        "yes please", "yes please do", "okay thanks",
    }
    no_replies = {
        "no", "no thanks", "nope", "nah", "not yet",
        "never mind", "stop", "don't book it",
    }

    if normalized in yes_replies:
        return "yes"
    if normalized in no_replies:
        return "no"

    if words and words[0] in {"yes", "y", "ok", "okay", "sure"}:
        return "unclear"
    if words and words[0] in {"no", "nope", "nah", "stop"}:
        return "no"

    return "unclear"


def cancellation_fee_cents(price_cents, start, now):
    """Fee in cents; raise ValueError if the appointment has already started."""
    remaining = start - now

    if remaining <= dt.timedelta(0):
        raise ValueError("Appointment has already started")

    if remaining >= dt.timedelta(hours=24):
        return 0
    if remaining >= dt.timedelta(hours=2):
        return price_cents // 2
    return price_cents


def make_tool(name, description, properties, required):
    return {
        "type": "function",
        "function": {
            "name": name,
            "description": description,
            "parameters": {
                "type": "object",
                "properties": properties,
                "required": required,
                "additionalProperties": False,
            },
        },
    }


TOOL_SCHEMAS = [
    make_tool(
        "list_services",
        "List the salon services and their prices.",
        {},
        [],
    ),
    make_tool(
        "find_slots",
        "Find available appointment slots for a service on a date.",
        {
            "service_id": {"type": "string"},
            "date": {"type": "string", "description": "Date in YYYY-MM-DD format"},
            "stylist_id": {"type": "string"},
        },
        ["service_id", "date"],
    ),
    make_tool(
        "get_my_bookings",
        "List the signed-in customer's bookings.",
        {},
        [],
    ),
    make_tool(
        "book_appointment",
        "Propose booking an appointment. Customer identity and approval are handled by the controller.",
        {
            "service_id": {"type": "string"},
            "stylist_id": {"type": "string"},
            "start": {"type": "string", "description": "Start time in YYYY-MM-DD HH:MM format"},
        },
        ["service_id", "stylist_id", "start"],
    ),
    make_tool(
        "reschedule_appointment",
        "Propose changing the time of one of the customer's bookings.",
        {
            "booking_id": {"type": "string", "description": "Booking ID such as B-1"},
            "new_start": {"type": "string", "description": "New start time in YYYY-MM-DD HH:MM format"},
        },
        ["booking_id", "new_start"],
    ),
    make_tool(
        "cancel_appointment",
        "Propose cancelling one of the customer's bookings.",
        {
            "booking_id": {"type": "string", "description": "Booking ID such as B-1"},
        },
        ["booking_id"],
    ),
    make_tool(
        "handoff_to_human",
        "Request help from a human staff member.",
        {
            "reason": {
                "type": "string",
                "enum": [
                    "customer_request",
                    "policy_exception",
                    "tool_failures",
                    "other",
                ],
            },
        },
        ["reason"],
    ),
]

class SalonAssistant:
    def __init__(self, model, backend, clock):
        self.model, self.backend, self.clock = model, backend, clock
        self.sessions, self.audit = {}, []

    def open_session(self, session_id, customer_id):
        self.sessions[session_id] = {
            "customer_id": customer_id,
            "messages": [],
            "pending": None,
            "evidence": [],
            "request_number": 0,
            "tool_call_ids": set(),
            "turn": 0,

        }

    def _record_user_event(self, session_id, turn, user_text):
        preview = re.sub(r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b", "[EMAIL]", user_text)
        preview = re.sub(r"(?<!\d)\+?\d{7,}(?!\d)", "[PHONE]", preview)
        preview = preview[:80]

        self.audit.append({
            "session": session_id,
            "turn": turn,
            "call_id": None,
            "tool": "(user)",
            "status": "received",
            "error_type": None,
            "attempts": 1,
            "preview": preview,
        })

    def _run_read_tool(self, session_id, turn, call_id, name, args):
        state = self.sessions[session_id]

        def record(status, error_type, data, attempts):
            preview = json.dumps(data, ensure_ascii=False, default=str)
            preview = re.sub(r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b", "[EMAIL]", preview)
            preview = re.sub(r"(?<!\d)\+?\d{7,}(?!\d)", "[PHONE]", preview)

            self.audit.append({
                "session": session_id,
                "turn": turn,
                "call_id": call_id,
                "tool": name,
                "status": status,
                "error_type": error_type,
                "attempts": attempts,
                "preview": preview[:80],
            })

            return {
                "call_id": call_id,
                "tool": name,
                "status": status,
                "error_type": error_type,
                "data": data,
            }

        if name not in {"list_services", "find_slots", "get_my_bookings"}:
            return record("error", "unknown_tool", None, 1)

        if not isinstance(args, dict):
            return record("error", "validation_error", None, 1)

        allowed = {
            "list_services": set(),
            "find_slots": {"service_id", "date", "stylist_id"},
            "get_my_bookings": set(),
        }
        required = {
            "list_services": set(),
            "find_slots": {"service_id", "date"},
            "get_my_bookings": set(),
        }

        if set(args) - allowed[name] or not required[name] <= set(args):
            return record("error", "validation_error", None, 1)
        if not all(isinstance(value, str) for value in args.values()):
            return record("error", "validation_error", None, 1)

        try:
            if name == "find_slots":
                dt.datetime.strptime(args["date"], "%Y-%m-%d")
                if "stylist_id" in args:
                    result = self.backend.free_slots(
                        args["service_id"], args["date"], args["stylist_id"]
                    )
                else:
                    result = self.backend.free_slots(
                        args["service_id"], args["date"]
                    )
            elif name == "list_services":
                result = self.backend.list_services()
            else:
                result = self.backend.bookings_for(state["customer_id"])

            return record("ok", None, result, 1)

        except BackendTimeout:
            try:
                if name == "find_slots":
                    if "stylist_id" in args:
                        result = self.backend.free_slots(
                            args["service_id"], args["date"], args["stylist_id"]
                        )
                    else:
                        result = self.backend.free_slots(
                            args["service_id"], args["date"]
                        )
                elif name == "list_services":
                    result = self.backend.list_services()
                else:
                    result = self.backend.bookings_for(state["customer_id"])

                return record("ok", None, result, 2)

            except BackendTimeout:
                return record("error", "timeout", None, 2)

        except (ValueError, KeyError, TypeError):
            return record("error", "validation_error", None, 1)

    def _propose_booking(self, session_id, turn, call_id, args):
        state = self.sessions[session_id]
        name = "book_appointment"

        def observation(status, error_type=None, data=None, attempts=1):
            preview = json.dumps(data, ensure_ascii=False, default=str)
            preview = re.sub(r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b", "[EMAIL]", preview)
            preview = re.sub(r"(?<!\d)\+?\d{7,}(?!\d)", "[PHONE]", preview)

            self.audit.append({
                "session": session_id,
                "turn": turn,
                "call_id": call_id,
                "tool": name,
                "status": status,
                "error_type": error_type,
                "attempts": attempts,
                "preview": preview[:80],
            })

            return {
                "call_id": call_id,
                "tool": name,
                "status": status,
                "error_type": error_type,
                "data": data,
            }

        if not isinstance(args, dict):
            return observation("error", "validation_error")

        if set(args) != {"service_id", "stylist_id", "start"}:
            return observation("error", "validation_error")

        if not all(isinstance(value, str) for value in args.values()):
            return observation("error", "validation_error")

        if state["pending"] is not None:
            return observation("blocked", "pending_exists")

        service_id = args["service_id"]
        stylist_id = args["stylist_id"]
        start = args["start"]

        try:
            start_dt = dt.datetime.strptime(start, "%Y-%m-%d %H:%M")
        except ValueError:
            return observation("error", "validation_error")

        if start_dt.minute not in {0, 30}:
            return observation("error", "validation_error")

        if service_id not in SERVICES or stylist_id not in STYLISTS:
            return observation("error", "validation_error")

        try:
            bookings = self.backend.bookings_for(state["customer_id"])
            slots = self.backend.free_slots(
                service_id, start_dt.strftime("%Y-%m-%d"), stylist_id
            )
        except BackendTimeout:
            try:
                bookings = self.backend.bookings_for(state["customer_id"])
                slots = self.backend.free_slots(
                    service_id, start_dt.strftime("%Y-%m-%d"), stylist_id
                )
            except BackendTimeout:
                return observation("error", "timeout", attempts=2)
        except (ValueError, KeyError):
            return observation("error", "validation_error")

        active_count = sum(b["status"] == "active" for b in bookings)
        if active_count >= MAX_ACTIVE:
            return observation("blocked", "limit_reached")

        requested = {
            "stylist_id": stylist_id,
            "start": start_dt.strftime("%Y-%m-%d %H:%M"),
        }
        if requested not in slots:
            alternatives = self.backend.free_slots(
                service_id, start_dt.strftime("%Y-%m-%d")
            )
            alternatives.sort(
                key=lambda slot: (
                    abs(
                        dt.datetime.strptime(slot["start"], "%Y-%m-%d %H:%M")
                        - start_dt
                    ),
                    slot["start"],
                    slot["stylist_id"],
                )
            )
            return observation(
                "blocked",
                "slot_unavailable",
                {"alternatives": alternatives[:3]},
            )

        service = SERVICES[service_id]
        stylist = STYLISTS[stylist_id]
        summary = (
            f"{service['name']} with {stylist['name']} at "
            f"{start_dt:%Y-%m-%d %H:%M} for "
            f"€{service['price_cents'] / 100:.2f}"
        )

        state["pending"] = {
            "action": "book",
            "service_id": service_id,
            "stylist_id": stylist_id,
            "start": start_dt.strftime("%Y-%m-%d %H:%M"),
        }

        return observation(
            "needs_confirmation",
            data={"summary": summary},
        )

    def _commit_booking(self, session_id, turn, call_id):
        state = self.sessions[session_id]
        pending = state["pending"]

        state["request_number"] += 1
        request_key = (
            f"{session_id}:{turn}:{state['request_number']}"
        )

        status = "error"
        error_type = None
        data = None
        attempts = 1

        def do_book():
            return self.backend.book(
                state["customer_id"],
                pending["service_id"],
                pending["stylist_id"],
                pending["start"],
                request_key,
            )

        try:
            data = do_book()
            status = "ok"

        except BackendTimeout:
            # شاید رزرو ثبت شده اما پاسخ گم شده باشد.
            data = self.backend.find_by_key(request_key)

            if data is not None:
                status = "ok"
            else:
                attempts = 2
                try:
                    data = do_book()
                    status = "ok"
                except BackendTimeout:
                    data = self.backend.find_by_key(request_key)
                    if data is not None:
                        status = "ok"
                    else:
                        status = "error"
                        error_type = "unknown_outcome"

        except SlotTaken:
            status = "error"
            error_type = "slot_taken"
            try:
                start_dt = dt.datetime.strptime(
                    pending["start"], "%Y-%m-%d %H:%M"
                )
                alternatives = self.backend.free_slots(
                    pending["service_id"],
                    start_dt.strftime("%Y-%m-%d"),
                )
                alternatives.sort(
                    key=lambda slot: (
                        abs(
                            dt.datetime.strptime(
                                slot["start"], "%Y-%m-%d %H:%M"
                            ) - start_dt
                        ),
                        slot["start"],
                        slot["stylist_id"],
                    )
                )
                data = {"alternatives": alternatives[:3]}
            except Exception:
                data = {"alternatives": []}

        except BackendError as exc:
            status = "blocked"
            error_type = exc.code

        preview = json.dumps(data, ensure_ascii=False, default=str)
        self.audit.append({
            "session": session_id,
            "turn": turn,
            "call_id": call_id,
            "tool": "book_appointment",
            "status": status,
            "error_type": error_type,
            "attempts": attempts,
            "preview": preview[:80],
        })

        return {
            "call_id": call_id,
            "tool": "book_appointment",
            "status": status,
            "error_type": error_type,
            "data": data,
        }

    def _propose_cancel(self, session_id, turn, call_id, args):
        state = self.sessions[session_id]

        def observation(status, error_type=None, data=None, attempts=1):
            preview = json.dumps(data, ensure_ascii=False, default=str)
            preview = re.sub(r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b", "[EMAIL]", preview)
            preview = re.sub(r"(?<!\d)\+?\d{7,}(?!\d)", "[PHONE]", preview)

            self.audit.append({
                "session": session_id,
                "turn": turn,
                "call_id": call_id,
                "tool": "cancel_appointment",
                "status": status,
                "error_type": error_type,
                "attempts": attempts,
                "preview": preview[:80],
            })

            return {
                "call_id": call_id,
                "tool": "cancel_appointment",
                "status": status,
                "error_type": error_type,
                "data": data,
            }

        if not isinstance(args, dict) or set(args) != {"booking_id"}:
            return observation("error", "validation_error")

        booking_id = args["booking_id"]
        if not isinstance(booking_id, str) or not re.fullmatch(r"B-\d+", booking_id):
            return observation("error", "validation_error")

        if state["pending"] is not None:
            return observation("blocked", "pending_exists")

        try:
            bookings = self.backend.bookings_for(state["customer_id"])
        except BackendTimeout:
            try:
                bookings = self.backend.bookings_for(state["customer_id"])
            except BackendTimeout:
                return observation("error", "timeout", attempts=2)

        booking = next(
            (
                b for b in bookings
                if b["booking_id"] == booking_id and b["status"] == "active"
            ),
            None,
        )
        if booking is None:
            return observation("not_found")

        try:
            fee_cents = cancellation_fee_cents(
                SERVICES[booking["service_id"]]["price_cents"],
                parse_dt(booking["start"]),
                self.clock(),
            )
        except ValueError:
            return observation("blocked", "already_started")

        summary = (
            f"Cancel {SERVICES[booking['service_id']]['name']} with "
            f"{STYLISTS[booking['stylist_id']]['name']} at {booking['start']}; "
            f"fee €{fee_cents / 100:.2f}"
        )

        state["pending"] = {
            "action": "cancel",
            "booking_id": booking_id,
            "fee_cents": fee_cents,
        }

        return observation(
    "needs_confirmation",
    data={
        "summary": summary,
        "fee_cents": fee_cents,
        "booking_id": booking_id,
    },
)

    def _commit_cancel(self, session_id, turn, call_id):
        state = self.sessions[session_id]
        pending = state["pending"]

        state["request_number"] += 1
        request_key = f"{session_id}:{turn}:{state['request_number']}"

        def do_cancel():
            return self.backend.cancel(
                state["customer_id"],
                pending["booking_id"],
                pending["fee_cents"],
                request_key,
            )

        status = "error"
        error_type = None
        data = None
        attempts = 1

        try:
            data = do_cancel()
            status = "ok"
        except BackendTimeout:
            data = self.backend.find_by_key(request_key)
            if data is not None:
                status = "ok"
            else:
                attempts = 2
                try:
                    data = do_cancel()
                    status = "ok"
                except BackendTimeout:
                    data = self.backend.find_by_key(request_key)
                    if data is not None:
                        status = "ok"
                    else:
                        error_type = "unknown_outcome"
        except BackendError as exc:
            error_type = exc.code

        preview = json.dumps(data, ensure_ascii=False, default=str)
        self.audit.append({
            "session": session_id,
            "turn": turn,
            "call_id": call_id,
            "tool": "cancel_appointment",
            "status": status,
            "error_type": error_type,
            "attempts": attempts,
            "preview": preview[:80],
        })

        return {
            "call_id": call_id,
            "tool": "cancel_appointment",
            "status": status,
            "error_type": error_type,
            "data": data,
        }

    def _propose_reschedule(self, session_id, turn, call_id, args):
        state = self.sessions[session_id]

        def observation(status, error_type=None, data=None, attempts=1):
            preview = json.dumps(data, ensure_ascii=False, default=str)
            self.audit.append({
                "session": session_id,
                "turn": turn,
                "call_id": call_id,
                "tool": "reschedule_appointment",
                "status": status,
                "error_type": error_type,
                "attempts": attempts,
                "preview": preview[:80],
            })
            return {
                "call_id": call_id,
                "tool": "reschedule_appointment",
                "status": status,
                "error_type": error_type,
                "data": data,
            }

        if not isinstance(args, dict) or set(args) != {"booking_id", "new_start"}:
            return observation("error", "validation_error")

        booking_id = args["booking_id"]
        new_start = args["new_start"]

        if (
            not isinstance(booking_id, str)
            or not re.fullmatch(r"B-\d+", booking_id)
            or not isinstance(new_start, str)
        ):
            return observation("error", "validation_error")

        if state["pending"] is not None:
            return observation("blocked", "pending_exists")

        try:
            new_dt = dt.datetime.strptime(new_start, "%Y-%m-%d %H:%M")
        except ValueError:
            return observation("error", "validation_error")

        if new_dt.minute not in {0, 30}:
            return observation("error", "validation_error")

        try:
            bookings = self.backend.bookings_for(state["customer_id"])
        except BackendTimeout:
            try:
                bookings = self.backend.bookings_for(state["customer_id"])
            except BackendTimeout:
                return observation("error", "timeout", attempts=2)

        booking = next(
            (
                b for b in bookings
                if b["booking_id"] == booking_id and b["status"] == "active"
            ),
            None,
        )
        if booking is None:
            return observation("not_found")

        old_start = parse_dt(booking["start"])
        if old_start - self.clock() < dt.timedelta(hours=24):
            return observation("blocked", "too_late")

        if booking["reschedules"] >= 1:
            return observation("blocked", "already_rescheduled")

        if new_dt == old_start:
            return observation("blocked", "same_slot")

        try:
            slots = self.backend.free_slots(
                booking["service_id"],
                new_dt.strftime("%Y-%m-%d"),
                booking["stylist_id"],
            )
        except BackendTimeout:
            try:
                slots = self.backend.free_slots(
                    booking["service_id"],
                    new_dt.strftime("%Y-%m-%d"),
                    booking["stylist_id"],
                )
            except BackendTimeout:
                return observation("error", "timeout", attempts=2)
        except (ValueError, KeyError):
            return observation("error", "validation_error")

        requested = {
            "stylist_id": booking["stylist_id"],
            "start": new_dt.strftime("%Y-%m-%d %H:%M"),
        }

        if requested not in slots:
            slots.sort(
                key=lambda slot: (
                    abs(
                        dt.datetime.strptime(
                            slot["start"], "%Y-%m-%d %H:%M"
                        ) - new_dt
                    ),
                    slot["start"],
                    slot["stylist_id"],
                )
            )
            return observation(
                "blocked",
                "slot_unavailable",
                {"alternatives": slots[:3]},
            )

        state["pending"] = {
            "action": "reschedule",
            "booking_id": booking_id,
            "new_start": new_dt.strftime("%Y-%m-%d %H:%M"),
        }

        summary = (
            f"Move {booking_id} from {booking['start']} "
            f"to {new_dt:%Y-%m-%d %H:%M}, with the same stylist and service"
        )
        return observation(
            "needs_confirmation",
            data={"summary": summary},
        )

    def _commit_reschedule(self, session_id, turn, call_id):
        state = self.sessions[session_id]
        pending = state["pending"]

        state["request_number"] += 1
        request_key = f"{session_id}:{turn}:{state['request_number']}"

        def do_reschedule():
            return self.backend.reschedule(
                state["customer_id"],
                pending["booking_id"],
                pending["new_start"],
                request_key,
            )

        status = "error"
        error_type = None
        data = None
        attempts = 1

        try:
            data = do_reschedule()
            status = "ok"
        except BackendTimeout:
            data = self.backend.find_by_key(request_key)
            if data is not None:
                status = "ok"
            else:
                attempts = 2
                try:
                    data = do_reschedule()
                    status = "ok"
                except BackendTimeout:
                    data = self.backend.find_by_key(request_key)
                    if data is not None:
                        status = "ok"
                    else:
                        error_type = "unknown_outcome"
        except SlotTaken:
            error_type = "slot_taken"
            try:
                booking = next(
                    b for b in self.backend.bookings_for(state["customer_id"])
                    if b["booking_id"] == pending["booking_id"]
                )
                date = pending["new_start"][:10]
                alternatives = self.backend.free_slots(
                    booking["service_id"], date, booking["stylist_id"]
                )
                data = {"alternatives": alternatives[:3]}
            except Exception:
                data = {"alternatives": []}
        except BackendError as exc:
            error_type = exc.code
            if exc.code in {
                "too_late", "already_rescheduled", "same_slot",
                "already_started", "slot_unavailable",
            }:
                status = "blocked"
            elif exc.code == "not_found":
                status = "not_found"

        preview = json.dumps(data, ensure_ascii=False, default=str)
        self.audit.append({
            "session": session_id,
            "turn": turn,
            "call_id": call_id,
            "tool": "reschedule_appointment",
            "status": status,
            "error_type": error_type,
            "attempts": attempts,
            "preview": preview[:80],
        })

        return {
            "call_id": call_id,
            "tool": "reschedule_appointment",
            "status": status,
            "error_type": error_type,
            "data": data,
        }

    def _run_handoff(self, session_id, turn, call_id, args):
        state = self.sessions[session_id]
        allowed_reasons = {
            "customer_request",
            "policy_exception",
            "tool_failures",
            "other",
        }

        def observation(status, error_type=None, data=None):
            preview = json.dumps(data, ensure_ascii=False, default=str)
            self.audit.append({
                "session": session_id,
                "turn": turn,
                "call_id": call_id,
                "tool": "handoff_to_human",
                "status": status,
                "error_type": error_type,
                "attempts": 1,
                "preview": preview[:80],
            })
            return {
                "call_id": call_id,
                "tool": "handoff_to_human",
                "status": status,
                "error_type": error_type,
                "data": data,
            }

        if (
            not isinstance(args, dict)
            or set(args) != {"reason"}
            or args["reason"] not in allowed_reasons
        ):
            return observation("error", "validation_error")

        last_user_text = next(
            (
                message["content"]
                for message in reversed(state["messages"])
                if message.get("role") == "user"
            ),
            "",
        )

        summary = re.sub(
            r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b",
            "[EMAIL]",
            last_user_text,
        )
        summary = re.sub(r"(?<!\d)\+?\d{7,}(?!\d)", "[PHONE]", summary)
        summary = summary[:300]

        try:
            ticket_id = self.backend.create_ticket({
                "customer_id": state["customer_id"],
                "reason": args["reason"],
                "summary": summary,
            })
        except Exception:
            return observation("error", "backend_error")

        return observation("ok", data={"ticket_id": ticket_id})

    def _ground_reply(self, session_id, turn, user_text, reply):
        state = self.sessions[session_id]

        evidence_parts = [
            user_text,
            json.dumps(state["evidence"], ensure_ascii=False, default=str),
        ]

        # قیمت‌ها و هزینه‌ها در دادهٔ backend به سنت هستند؛
        # آن‌ها را هم به شکل یورو به شواهد اضافه می‌کنیم.
        def add_money_evidence(value):
            if isinstance(value, dict):
                for key, item in value.items():
                    if (
                        key in {"price_cents", "fee_cents"}
                        and isinstance(item, int)
                        and not isinstance(item, bool)
                    ):
                        evidence_parts.append(f"€{item / 100:.2f}")
                    add_money_evidence(item)
            elif isinstance(value, list):
                for item in value:
                    add_money_evidence(item)

        add_money_evidence(state["evidence"])
        evidence = " ".join(evidence_parts)

        # هر ساعت ذکرشده باید در شواهد باشد.
        reply_times = re.findall(
            r"\b(?:[01]\d|2[0-3]):[0-5]\d\b", reply
        )
        evidence_times = set(
            re.findall(r"\b(?:[01]\d|2[0-3]):[0-5]\d\b", evidence)
        )
        if any(value not in evidence_times for value in reply_times):
            return SAFE_FALLBACK

        # هر مبلغ یورویی باید در شواهد باشد.
        def euro_amounts(text):
            amounts = set()
            for value in re.findall(
                r"€\s*(\d+(?:[.,]\d{1,2})?)", text
            ):
                normalized = value.replace(",", ".")
                amounts.add(round(float(normalized) * 100))
            return amounts

        if not euro_amounts(reply) <= euro_amounts(evidence):
            return SAFE_FALLBACK

        # هر شناسهٔ رزرو باید در شواهد باشد.
        reply_ids = {
            value.upper()
            for value in re.findall(r"\bB-\d+\b", reply, flags=re.IGNORECASE)
        }
        evidence_ids = {
            value.upper()
            for value in re.findall(r"\bB-\d+\b", evidence, flags=re.IGNORECASE)
        }
        if not reply_ids <= evidence_ids:
            return SAFE_FALLBACK

        # ادعای موفقیت فقط با تغییر موفق همین نوبت مجاز است.
        claims_success = re.search(
            r"\b(?:is|was|has been|have been)\s+"
            r"(?:booked|cancelled|canceled|rescheduled)\b"
            r"|\b(?:booked|cancelled|canceled|rescheduled)"
            r"\s+successfully\b",
            reply,
            flags=re.IGNORECASE,
        )
        if claims_success:
            write_tools = {
                "book_appointment",
                "cancel_appointment",
                "reschedule_appointment",
            }
            successful_write = any(
                event["session"] == session_id
                and event["turn"] == turn
                and event["tool"] in write_tools
                and event["status"] == "ok"
                for event in self.audit
            )
            if not successful_write:
                return SAFE_FALLBACK

        return reply

    def handle(self, session_id, user_text):
        state = self.sessions[session_id]
        state["turn"] += 1
        turn = state["turn"]

        self._record_user_event(session_id, turn, user_text)

        now = self.clock()
        system_message = {
            "role": "system",
            "content": (
                f"Now: {now:%Y-%m-%d %H:%M} ({now:%A})\n"
                "You are Luna, the salon assistant. "
                "Use tools to check salon information. "
                "Never claim a booking changed unless the controller confirms it."
            ),
        }

        messages = state["messages"]
        if messages and messages[0]["role"] == "system":
            messages[0] = system_message
        else:
            messages.insert(0, system_message)

        messages.append({"role": "user", "content": user_text})

        def fit_history():
            while len(messages) > 24:
                next_user = next(
                    (
                        i for i in range(2, len(messages))
                        if messages[i].get("role") == "user"
                    ),
                    None,
                )
                if next_user is None:
                    break
                del messages[1:next_user]

        pending = state["pending"]
        if pending is not None:
            decision = interpret_reply(user_text)

            if decision == "yes" and pending["action"] in {
                "book", "cancel", "reschedule"
            }:
                call_id = f"confirm-{session_id}-{turn}"
                while call_id in state["tool_call_ids"]:
                    call_id += "-new"
                state["tool_call_ids"].add(call_id)

                if pending["action"] == "book":
                    tool_name = "book_appointment"
                    args = {
                        "service_id": pending["service_id"],
                        "stylist_id": pending["stylist_id"],
                        "start": pending["start"],
                    }
                elif pending["action"] == "cancel":
                    tool_name = "cancel_appointment"
                    args = {"booking_id": pending["booking_id"]}
                else:
                    tool_name = "reschedule_appointment"
                    args = {
                        "booking_id": pending["booking_id"],
                        "new_start": pending["new_start"],
                    }

                messages.append({
                    "role": "assistant",
                    "content": None,
                    "tool_calls": [{
                        "id": call_id,
                        "type": "function",
                        "function": {
                            "name": tool_name,
                            "arguments": json.dumps(args),
                        },
                    }],
                })

                if pending["action"] == "book":
                    observation = self._commit_booking(
                        session_id, turn, call_id
                    )
                elif pending["action"] == "cancel":
                    observation = self._commit_cancel(
                        session_id, turn, call_id
                    )
                else:
                    observation = self._commit_reschedule(
                        session_id, turn, call_id
                    )

                state["pending"] = None
                state["evidence"].append(observation)

                messages.append({
                    "role": "tool",
                    "tool_call_id": call_id,
                    "content": json.dumps(observation, ensure_ascii=False),
                })

            else:
                state["pending"] = None

        model_calls = 0
        tool_calls_count = 0
        serious_failures = 0
        handoff_created = False

        while model_calls < 6:
            fit_history()
            result = self.model(messages, TOOL_SCHEMAS)
            model_calls += 1

            if isinstance(result, dict) and isinstance(result.get("final"), str):
                reply = self._ground_reply(
                    session_id, turn, user_text, result["final"]
                )
                messages.append({"role": "assistant", "content": reply})
                return reply

            if not isinstance(result, dict):
                return SAFE_FALLBACK

            calls = result.get("tool_calls")
            if (
                not isinstance(calls, list)
                or not calls
                or tool_calls_count + len(calls) > 8
            ):
                return SAFE_FALLBACK

            parsed_calls = []
            new_ids = set()

            for call in calls:
                if not isinstance(call, dict):
                    return SAFE_FALLBACK

                function = call.get("function", call)
                if not isinstance(function, dict):
                    return SAFE_FALLBACK

                call_id = call.get("id")
                name = function.get("name")
                raw_args = function.get("arguments")

                if not isinstance(call_id, str) or not call_id:
                    return SAFE_FALLBACK
                if call_id in state["tool_call_ids"] or call_id in new_ids:
                    return SAFE_FALLBACK
                if not isinstance(name, str):
                    return SAFE_FALLBACK

                try:
                    args = (
                        json.loads(raw_args)
                        if isinstance(raw_args, str)
                        else raw_args
                    )
                except (json.JSONDecodeError, TypeError):
                    return SAFE_FALLBACK

                if not isinstance(args, dict):
                    return SAFE_FALLBACK

                new_ids.add(call_id)
                parsed_calls.append((call_id, name, args))

            state["tool_call_ids"].update(new_ids)
            tool_calls_count += len(parsed_calls)

            assistant_calls = []
            for call_id, name, args in parsed_calls:
                assistant_calls.append({
                    "id": call_id,
                    "type": "function",
                    "function": {
                        "name": name,
                        "arguments": json.dumps(args),
                    },
                })

            messages.append({
                "role": "assistant",
                "content": None,
                "tool_calls": assistant_calls,
            })

            for call_id, name, args in parsed_calls:
                if name == "book_appointment":
                    observation = self._propose_booking(
                        session_id, turn, call_id, args
                    )
                elif name == "cancel_appointment":
                    observation = self._propose_cancel(
                        session_id, turn, call_id, args
                    )
                elif name == "reschedule_appointment":
                    observation = self._propose_reschedule(
                        session_id, turn, call_id, args
                    )
                elif name == "handoff_to_human":
                    observation = self._run_handoff(
                        session_id, turn, call_id, args
                    )
                else:
                    observation = self._run_read_tool(
                        session_id, turn, call_id, name, args
                    )

                state["evidence"].append(observation)

                messages.append({
                    "role": "tool",
                    "tool_call_id": call_id,
                    "content": json.dumps(observation, ensure_ascii=False),
                })

                if (
                    observation.get("status") == "error"
                    and observation.get("error_type")
                    in {"timeout", "unknown_outcome"}
                ):
                    serious_failures += 1

            if serious_failures >= 2 and not handoff_created:
                auto_call_id = f"auto-handoff-{session_id}-{turn}"
                self._run_handoff(
                    session_id,
                    turn,
                    auto_call_id,
                    {"reason": "tool_failures"},
                )
                handoff_created = True

        return SAFE_FALLBACK
```

<h1>بخش C — اجرای Luna با یک مدل واقعی (۱۵ امتیاز)</h1>
<p>مدل‌های اسکریپت‌شده همیشه طبق سناریوی ازپیش‌تعیین‌شده رفتار می‌کنند، اما مدل‌های واقعی ممکن است پاسخ‌های پیش‌بینی‌نشده بدهند. در این بخش یک <code>LLM</code> واقعی را به <strong>همان</strong> <code>SalonAssistant</code> متصل می‌کنید. نباید برای این کار چیزی در منطق کنترلر تغییر دهید. هدف این است که ببینید کدام بخش‌ها در استفادهٔ واقعی مشکل پیدا می‌کنند.</p>
<p>تنظیمات مانند روش معمول دوره است: <code>OpenAI SDK</code>، مقدار <code>LLM_URL</code> و کلید دسترسی از متغیر محیطی یا <code>getpass</code>. <strong>کلید را هرگز چاپ، ذخیره یا با دیگران به اشتراک نگذارید.</strong> کد <code>adapter</code> زیر پیام‌های شما را به قالب چت OpenAI تبدیل می‌کند.</p>
<p><strong>مواردی که باید تحویل دهید</strong> (در بخش گزارش انتهای Notebook): حداقل <strong>شش گفتگو</strong> با مدل واقعی اجرا کنید. می‌توانید از شخصیت‌های پیشنهادی زیر یا موارد خودتان استفاده کنید. متن گفتگوها را نگه دارید و برای هر مورد توضیح دهید <strong>Luna چه کاری را درست یا اشتباه انجام داده است</strong>. باید حداقل <strong>دو باگ یا ضعف واقعی</strong> را که فقط در کار با مدل واقعی مشخص شده‌اند پیدا کنید، توضیح دهید و <strong>برطرف کنید</strong>؛ این ضعف‌ها می‌توانند در کنترلر یا پیام <code>system</code> باشند. گزارشی که فقط بگوید «همه‌چیز درست کار کرد» مانند این است که بخش اجرای واقعی انجام نشده باشد.</p>

```python
# GIVEN — do not edit this cell (the grader uses its own copy)
import os
from getpass import getpass

MODEL_NAME = 'google/gemma-4-26b-a4b-it'

def to_openai(messages):
    """Our messages -> OpenAI chat format (tool calls may be {'id','name','arguments'} or already OpenAI-shaped)."""
    out = []
    for m in messages:
        if m['role'] == 'assistant' and m.get('tool_calls'):
            calls_out = []
            for c in m['tool_calls']:
                if 'function' in c:
                    calls_out.append(c); continue
                args = c['arguments'] if isinstance(c['arguments'], str) else json.dumps(c['arguments'])
                calls_out.append({'id': c['id'], 'type': 'function', 'function': {'name': c['name'], 'arguments': args}})
            out.append({'role': 'assistant', 'content': m.get('content'), 'tool_calls': calls_out})
        else:
            out.append({k: v for k, v in m.items() if k in ('role', 'content', 'tool_call_id')})
    return out

def make_llm_model(client, model_name=MODEL_NAME):
    """Wrap an OpenAI-compatible client as `model(messages, tools)` for SalonAssistant."""
    def model(messages, tools=None):
        response = client.chat.completions.create(model=model_name, messages=to_openai(messages), tools=tools or None, temperature=0)
        message = response.choices[0].message
        if getattr(message, 'tool_calls', None):
            return {'tool_calls': [{'id': tc.id, 'name': tc.function.name, 'arguments': tc.function.arguments} for tc in message.tool_calls]}
        return {'final': message.content or ''}
    return model

LIVE = False
api_key = os.environ.get('OPENROUTER_API_KEY') or getpass('OpenRouter API key (hidden; press Enter to skip the live part): ')
if api_key:
    try:
        from openai import OpenAI
        client = OpenAI(base_url=os.environ.get('LLM_URL', 'https://llm-proxy.op8.ir/openrouter'), api_key=api_key, timeout=45.0, max_retries=0)
        live_model = make_llm_model(client)
        LIVE = True
        print('Live model ready:', MODEL_NAME)
    except Exception as exc:
        print('Could not create the client:', type(exc).__name__)
else:
    print('No key: live part skipped.')
```

```python
def converse(assistant, session_id, first_message, customer_id='c_ada', max_turns=12):
    """Interactive chat with Luna. Type an empty line to stop. Prints the audit trail of tool calls after each turn."""
    if session_id not in assistant.sessions:
        assistant.open_session(session_id, customer_id)
    text, seen = first_message, 0
    for _ in range(max_turns):
        print('\nYOU :', text)
        print('LUNA:', assistant.handle(session_id, text))
        for event in assistant.audit[seen:]:
            if event['tool'] != '(user)':
                print('   · tool', event['tool'], event['status'], event['error_type'] or '', f"(attempts {event['attempts']})")
        seen = len(assistant.audit)
        text = input('\nYOU > ').strip()
        if not text: break

if LIVE:
    live_backend = SalonBackend()
    luna = SalonAssistant(live_model, live_backend, lambda: live_backend.now)
    converse(luna, 'live-1', "Hi, I'd like a haircut this Thursday afternoon.")
else:
    print('Live run skipped (no key). You still have to do it for the grade: ask your instructor for access.')
```
