# Question 1 `GYM members`

```python
"""GYM members manager system"""

# imports
import numpy as np

# gym data nd array
gym_data = np.array(
    [
        # zero values == Null
        # column 0 == age
        # column 1 == weight
        # column 2 == height
        # column 3 == Number of sessions per week
        [28, 75, 175, 4],
        [34, 68, 168, 3],
        [45, 82, 180, 2],
        [22, 58, 162, 5],
        [38, 90, 0, 1],
        [29, 65, 170, 0],
    ],
    dtype=float,
)
print(f"Before preprocessing:\n{gym_data}")
# members name
members_name = np.array(["Ali", "Sara", "Reza", "Neda", "Hassan", "Maryam"])

# preprocessing
columns_mean = np.mean(gym_data, axis=0)
# condition = where cells are zero
# np.take = replace column mean  with zero cell index
# np.where return row and col and [1] tell only return column
gym_data[gym_data == 0] = np.take(columns_mean, np.where(gym_data == 0)[1])
print(f"\nafter preprocessing:\n{gym_data}")

# fitness array weight , height , sessions per week featurs
fitness_arr = gym_data[:, 1:4]
# height **2 >> meter | weight >> kilogram
bmi_score = fitness_arr[:, 0] / ((fitness_arr[:, 1] * 0.01) ** 2)
# fitness score a new feature in fitness array == (bmi score* w1)*(sessions * w2)
weights = np.array([0.7, 0.3])
fitness_score = ((bmi_score * weights[0]) + (fitness_arr[:, -1] * weights[1])).reshape(
    -1, 1
)
# stacking arrays
fitness_arr = np.hstack((fitness_arr, fitness_score))
# fitness array with a new feature
print(f"\nfitness table:\n{fitness_arr}")
# Who is the fittest guy in the gym ?
fittest_index = fitness_arr[:, -1].argmax()
print(f"\nfittest member :{members_name[fittest_index]}\
          with score: {fitness_arr[:,-1].max():.2f}")

# use std to find the record that has the most different session per week
# to detect anomaly in sessions we should measure distance
#  from mean and we can devide it by std == Z score
# bigger Z score == more distance
fit_std = fitness_arr[:, 2].std()
fit_mean = fitness_arr[:, 2].mean()
z_score = np.abs((fitness_arr[:, 2] - fit_mean) / fit_std)
anomaly_index = z_score.argmax()
anomaly_person = members_name[anomaly_index]
print(f"\nanomaly person in session col is {anomaly_person},\
      {fitness_arr[anomaly_index,2]} session per week")
```

# Question 2 `Food offer`

```python
#This cell will Sugest food to customers by their taste

# imports
import numpy as np

# variables
recipes = np.array(
    [
        # first column =  Cooking time
        # second column = Calories
        # third column = spiciness
        # forth column = nb of ingredients
        [15, 350, 2, 5],
        [45, 600, 7, 10],
        [10, 200, 0, 3],
        [30, 450, 5, 7],
        [60, 800, 8, 12],
    ]
)
# Food's names
recipe_names = np.array(["Salad", "Curry", "Toast", "Pasta", "Stew"])

# shape and view of data
print(f"shape of recipes array:\n{recipes.shape}\nview of recipes array:\n{recipes}")

# preprocessing | Min Max scaling
min_arr = np.min(recipes, axis=0)
max_arr = np.max(recipes, axis=0)
scaled_recipes = (recipes - min_arr) / (max_arr - min_arr)

print(f"\nrecipe array scaled:\n{scaled_recipes}")

# test data
users = np.array([[10, 250, 1, 4], [50, 700, 8, 11], [25, 400, 4, 6]])
# scale user with recipes array min and max
scaled_customer = (users - min_arr) / (max_arr - min_arr)
print(f"\ncustomer array scaled:\n{scaled_customer}")
# scaled_customer.reshape(3,1,4) = 3,1,4 that means our 1 dim
# in each dim has one person and four features that will be compared to 5 food with 4 features
# in the other hand numpy said 1 in N is ok , so we have 3 of 1 in N
# So each row has a 1×4 matrix that is OK to broadcasting
users_taste = np.linalg.norm(scaled_customer.reshape(3, 1, 4) - scaled_recipes, axis=2)
# find favorit food
taste_index = users_taste.argmin(axis=1)
user_sorted_taste = users_taste.argsort(axis=1)
print(f"favorit foods by order : {recipe_names[taste_index]}")
print(
    f"\nfavorit foods by order (first food = nearest food): \n{recipe_names[user_sorted_taste]}"
)
```

# Question 3 `High School`

```python
#This program is a high school grading management system

# imports
import numpy as np

# variables
scores = np.array(
    [
        # first feature = Quiz
        # second feature = Midterm
        # third feature = Final exam
        [18, 15, 20],
        [12, 14, 16],
        [20, 19, 18],
        [10, 8, 15],
    ]
)

print(f"scores shape:\n{scores.shape}")
# weight schemes
weight_schemes = np.array(
    [
        # scheme A | Quiz = most important
        [0.5, 0.3, 0.2],
        # scheme B | final = most important
        [0.2, 0.3, 0.5],
        # scheme C | final = very very important!
        [0.1, 0.2, 0.7],
    ]
)

print(f"weight schemes shape:\n{weight_schemes.shape}")
# final score each row has student score with A,B,C schemes.
final_scores = scores @ weight_schemes.T
print(f"each row is a student each column is a weight scheme:\n{final_scores}\n\
final score array shape = {final_scores.shape}\n")

# whitch scheme for each student has most score
scheme_map = np.array(["A", "B", "C"])
max_score_each = final_scores.argmax(axis=1)
best_scheme_each = scheme_map[max_score_each]
for i, scheme in enumerate(best_scheme_each, start=1):
    print(f"student {i},best scheme is {scheme}")

# best scheme for all classes
mean_score_all = final_scores.mean(axis=0)
max_score_all = mean_score_all.argmax()

print(
    f"{scheme_map[max_score_all]} is better for all students:\n\
{final_scores[:,max_score_all]} are students \
final score with scheme {scheme_map[max_score_all]}"
)
print("#-#" * 10)
# scheme D is new scheme [0.2, 0.4 , 0.4]
weight_schemes_n = np.array(
    [
        # scheme A | Quiz = most important
        [0.5, 0.3, 0.2],
        # scheme B | final = most important
        [0.2, 0.3, 0.5],
        # scheme C | final = very very important!
        [0.1, 0.2, 0.7],
        # scheme D | final = Quiz × 2
        [0.2, 0.4, 0.4],
    ]
)

print(f"weight schemes shape:\n{weight_schemes_n.shape}")
# final score each row has student score with A,B,C,D schemes.
final_scores_n = scores @ weight_schemes_n.T
print(f"each row is a student each column is a weight scheme:\n{final_scores_n}\n\
final score array shape = {final_scores_n.shape}\n")

scheme_map_n = np.array(["A", "B", "C", "D"])

# whitch scheme for each student has most score
max_score_each_n = final_scores_n.argmax(axis=1)
best_scheme_each_n = scheme_map_n[max_score_each_n]
for i, scheme in enumerate(best_scheme_each_n, start=1):
    print(f"student {i},best scheme is {scheme}")

# best scheme for all classes
mean_score_all_n = final_scores_n.mean(axis=0)
max_score_all_n = mean_score_all_n.argmax()

print(
    f"{scheme_map_n[max_score_all_n]} is better for all students:\n\
{final_scores_n[:,max_score_all_n]} are students \
final score with scheme {scheme_map_n[max_score_all_n]}"
)
```
